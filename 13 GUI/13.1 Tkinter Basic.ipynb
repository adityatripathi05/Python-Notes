## GUI (Graphical User Interface)
- A GUI is a system of interactive components such as icons and other graphical objects that help a user to interact with computer software and other applications more easily.


## Tcl (Tool Command Language)/Tk(Toolkit)
- Tcl is a general purpose multi-paradigm system programming language.
    - It is a scripting language that aims at providing the ability for applications to communicate with each other.
    - It is suitable for a very wide range of uses, including web and desktop applications, networking, administration, testing and many more.
- Tk is a cross platform widget toolkit used for building GUI in many languages.


## Tkinter Library
- A computer system consist of **Hardware**.
- This computer system hardware is controlled by an **Operating System**.
<br/><br/>
- High level programming languages are on top of this Operating System layer.
    - <img src='./Image/13.1 Image a.png'>
    
    - i.e, when we write our code in python to control our hardware, python do so with the help of Operating System.
<br/><br/>
- If we wish to built a GUI using python, then just having O.S and Hardware is insufficient.
    - Because to build a GUI, python require many other services and one such is **Tcl/Tk**.
    - <img src='./Image/13.1 Image b.png'>
    
<br/><br/>
- And since Tcl/Tk is a different programming language so we can't control the services of it using python syntax alone. To deal with this another package in python is introduced called **tkinter**.
    - <img src='./Image/13.1 Image c.png'>
    
    - tkinter act as an intermediate layer between python and Tcl/Tk. hence allow python to use services provided by Tcl/Tk to build GUI.
- The tkinter package ("Tk interface") is the standard Python interface to the Tk GUI toolkit. It is a thin object-oriented layer on top of Tcl/Tk.
<br/><br/>
- Other Open Source libraries for GUI are WxPython, PyQt5, Jython etc.

### Geometry Management:
- The Tkinter geometry specifies the method by using which, the widgets are represented on display.
- The python Tkinter provides the following geometry methods:
    - **1. pack()** 
    - **2. grid()** 
    - **3. place()**

#### pack():
- It organizes widgets in blocks before placing them in the parent widget.
- Syntax: `widget.pack(options)` 
- A list of possible options that can be passed in pack() is given below:
    - **expand:** If the expand is set to **True**, the widget expands to fill any space.
    - **fill:** By default, the fill is set to **NONE**. The value controls how to fill the space.
        - **tk.BOTH** means that the widget should expand both horisontally and vertically, 
        - **tk.X** means that it should expand only horisontally, and 
        - **tk.Y** means that it should expand only vertically.
    - **side:** It represents the side of the interface to which the widget is to be placed on the window. 
        - **tk.LEFT**, **tk.RIGHT**, **tk.TOP**, **tk.BOTTOM**
<br/><br/>
- **When to use the Pack Manager?**
    - Compared to the grid manager, the pack manager is somewhat limited, but it’s much easier to use in a few, but quite common situations:
        - 1. Put a widget inside a frame (or any other container widget), and have it fill the entire frame
        - 2. Place a number of widgets on top of each other
        - 3. Place a number of widgets side by side
<br/><br/>        
- **NOTE:** Don't mix grid and pack in the same master window.

```python
import tkinter as tk
win = tk.Tk()  
win.title('My GUI') 
win.geometry("300x300")

tk.Label(win, text="First").pack() # place at Top centre bydefault

win.mainloop()
```

#### grid():
- grid() geometry manager organizes the widgets in the tabular form. 
    - This is a more organized way to place the widgets to the python application.
- **Syntax:** `widget.grid(options)`  
- A list of possible options that can be passed inside the grid() method is given below:
    - **column:** The column number in which the widget is to be placed. 
        - The leftmost column is represented by 0.
    - **columnspan:** The width of the widget. 
        - It represents the number of columns up to which, the column is expanded.
    - **ipadx, ipady:** It represents the number of pixels to **pad the widget inside** the widget's border.
    - **padx, pady:** It represents the number of pixels to **pad the widget outside** the widget's border.
    - **row:** The row number in which the widget is to be placed. 
        - The topmost row is represented by 0.
    - **rowspan:** The height of the widget.
        - The number of the row up to which the widget is expanded.
    - **Sticky:** If the cell is larger than a widget, then sticky is used to specify the position of the widget inside the cell. 
        - It's value may be **'N'**, **'E'**, **'W'**, **'S'**, **'NE'**, **'NW'**, **'NS'**, **'EW'**, **'ES'**.

```python
import tkinter as tk
win = tk.Tk()  
win.title('My GUI') 
win.geometry("300x300")   

tk.Label(win, text="First").grid(row=0,column=0)
tk.Label(win, text="Second").grid(row=1,column=1)

win.mainloop()
```

#### place():
- The place() geometry manager organizes the widgets to the specific x and y coordinates.
- **Syntax:** `widget.place(options)`  
- A list of possible options that can be passed inside the place() method is given below:
    - **anchor:** It represents the exact position of the widget within the container. 
        - The default value (direction) is NW (the upper left corner)
    - **bordermode:** The default value of the border type is INSIDE that refers to ignore the parent's inside the border. 
        - The other option is OUTSIDE.
    - **height, width:** It refers to the height and width in pixels.
    - **relheight, relwidth:** It is represented as the float between 0.0 and 1.0 indicating the fraction of the parent's height and width.
    - **relx, rely:** It is represented as the float between 0.0 and 1.0 that is the offset in the horizontal and vertical direction.
    - **x, y:** It refers to the horizontal and vertical offset in the pixels.

```python
import tkinter as tk
win = tk.Tk()  
win.title('My GUI') 
win.geometry("300x300")   

tk.Label(win, text="First").place(x=0,y=0)
tk.Label(win, text="Second").place(x=20,y=20)

win.mainloop()
```

### Add Entry Box Widget:
- The Entry widget is used to display a single-line text field for accepting values from a user.
- The kinds of variables subclassed from a class called Variable, defined in the Tkinter module.
    - **x = StringVar()**: Holds a string; default value `""`
    - **x = IntVar()**: Holds an integer; default value 0
    - **x = DoubleVar()**: Holds a float; default value 0.0
    - **x = BooleanVar()**: Holds a boolean, returns 0 for False and 1 for True
- To **read the current value** of such a variable, call **get()** method. 
- To **change the value** of such a variable, call **set()** method.

```python
import tkinter as tk
win= tk.Tk()
win.title('My GUI')
win.geometry('300x300')
win.configure(background='#33FFBE')
label=tk.Label(win,text='Name', font=('Ariel',10))
label.grid(row=0, column=0,padx=2)

var=tk.StringVar()
entry=tk.Entry(win,width=15,textvariable=var)
entry.grid(row=0,column=1,padx=2)

win.mainloop()
```

### Add Radiobutton Widget:

```python
import tkinter as tk
win= tk.Tk()
win.title('My GUI')
win.geometry('300x300')

labels=['Name','Age','Email','Gender','Country','Weight']
for i in range(len(labels)):
    label='label'+str(i)
    label=tk.Label(win,text=labels[i])
    label.grid(row=i, column=0,sticky='E',padx=2,pady=2)

var={'var0':tk.StringVar(),
     'var1':tk.StringVar(),
     'var2':tk.StringVar()}
entry=['entry0','entry1','entry2']
for i in range(len(var)):
    entry[i]=tk.Entry(win,width=15,
                       textvariable=var['var'+str(i)])
    entry[i].grid(row=i,column=1,padx=2,pady=2)
entry[0].focus()

var3=tk.StringVar()
txt=['Male','Female','Other']
for i in range(3):
    rdbtn=f'rdbtn{i}'
    rdbtn=tk.Radiobutton(win,text=txt[i],value=txt[i],
                          variable=var3)
    rdbtn.grid(row=3,column=1+i, sticky='E')
var3.set(0)  # Any arbitrary value to disselect

win.mainloop()
```

### Add Combobox Widget:
- Combobox in tkinter is a drop down list, which can hold multiple values and show one item at a time. 
- Combobox is useful for select one option in many choice. 
- Combobox widget is a class of ttk module of tkinter library.

```python
import tkinter as tk
import tkinter.ttk as ttk
win= tk.Tk()
win.title('My GUI')
win.geometry('300x300')

labels=['Name','Age','Email','Gender','Country','Weight']
for i in range(len(labels)):
    label='label'+str(i)
    label=tk.Label(win,text=labels[i])
    label.grid(row=i, column=0,sticky='E',padx=2,pady=2)

var={'var0':tk.StringVar(),
     'var1':tk.StringVar(),
     'var2':tk.StringVar()}
entry=['entry0','entry1','entry2']
for i in range(len(var)):
    entry[i]=tk.Entry(win,width=15,
                       textvariable=var['var'+str(i)])
    entry[i].grid(row=i,column=1,padx=2,pady=2)
entry[0].focus()

var3=tk.StringVar()
txt=['Male','Female','Other']
for i in range(3):
    rdbtn=f'rdbtn{i}'
    rdbtn=tk.Radiobutton(win,text=txt[i],value=txt[i], variable=var3)
    rdbtn.grid(row=3,column=1+i, sticky='E')
var3.set(0)

var4=tk.StringVar()
cbbox=ttk.Combobox(win,width=13,textvariable=var4, state='readonly')
cbbox['value']=['India','America','Canada','Brazil']
cbbox.current(0)
cbbox.grid(row=4,column=1)
    
win.mainloop()
```

### Add Button Widget:
- The button widget is used for user interaction. 
    - We can associate a method or function with a button which is called when the button is pressed.
- Python allows us to configure the look of the button according to our requirements. 
    - Various options can be set or reset depending upon the requirements.

### Check Prime

```python
import tkinter as tk
class CountPrime:
    def __init__(self):
        self.win = tk.Tk()
        self.win.title("Count Prime")
        self.label=tk.Label(self.win,text='Enter no. of Elements: ')
        self.label.grid(row=0,column=0)
        self.freq_entry=tk.Entry(self.win)
        self.freq_entry.bind("<Return>", self.value_entry)
        self.freq_entry.grid(row=0,column=1)
        self.win.mainloop()
        
    def value_entry(self,event):
        self.variables={}
        for v in range(int(self.freq_entry.get())):
            self.variables[f'var{v}']=tk.StringVar()
        for self.i in range(int(self.freq_entry.get())):
            self.label=tk.Label(self.win,text=f'Enter value {self.i+1}: ')
            self.label.grid(row=self.i+1,column=0)
            self.value_entry=tk.Entry(self.win,textvariable=self.variables[f'var{self.i}'])
            self.value_entry.grid(row=self.i+1,column=1)
        self.check_btn=tk.Button(self.win,text='Check',command=self.check_prime)
        self.check_btn.grid(row=self.i+2,column=1)
        self.result = tk.Label(self.win)
        self.result.grid(row=self.i+4,column=0, sticky='E')
        
    def check_prime(self):
        self.list_=[]
        self.prime=[]
        for k in self.variables.keys():
            self.list_.append(int(self.variables[k].get()))
        for num in self.list_:
            c=0
            for n in range(1, num+1):
                if num%n==0:
                    c += 1
            if c==2:
                self.prime.append(str(num))
        self.result.configure(text="Result: " + ','.join(self.prime)+" are Prime")
            
CountPrime()
```

### Autocomplete Text:

```python
import tkinter as tk

def on_keyrelease(event):

    # get text from entry
    value = event.widget.get()
    value = value.strip().lower()

    # get data from test_list
    if value == '':
        data = test_list
    else:
        data = []
        for item in test_list:
            if value in item.lower():
                data.append(item)                

    # update data in listbox
    listbox_update(data)


def listbox_update(data):
    # delete previous data
    listbox.delete(0, 'end')

    # sorting data 
    data = sorted(data, key=str.lower)

    # put new data
    for item in data:
        listbox.insert('end', item)


def on_select(event):
    # display element selected on list
    print('(event) previous:', event.widget.get('active'))
    print('(event)  current:', event.widget.get(event.widget.curselection()))
    print('---')


# --- main ---

test_list = ('apple', 'banana', 'Cranberry', 'dogwood', 'alpha', 'Acorn', 'Anise', 'Strawberry' )

root = tk.Tk()

entry = tk.Entry(root)
entry.pack()
entry.bind('<KeyRelease>', on_keyrelease)

listbox = tk.Listbox(root)
listbox.pack()
#listbox.bind('<Double-Button-1>', on_select)
listbox.bind('<<ListboxSelect>>', on_select)
listbox_update(test_list)

root.mainloop()
```

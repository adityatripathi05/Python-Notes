- A more elegant way of automatically iterating is by using the for loop. Using this, we can iterate over any object that can return an iterator.


## Iteration Vs Iterators Vs Iterables:
### Iteration:
- It is the process of taking one element at a time from a sequence of elements i.e, Any time you use a loop, to go over a group of items is iteration.


### Iterators:
- Iterators are objects that can be iterated upon.
    - i.e, An object which will return data, one element at a time.
- We use the next() function to manually iterate through all the items of an iterator. 
- When we reach the end and there is no more data to be returned, it will raise StopIteration


### Iterables:
- An object is called iterable if we can get an iterator from it.
- The iter() function (which in turn calls the __iter__() method) returns an iterator from them.
- So an iterable is an object that you can get an iterator from.
- Most of built-in containers in Python like: list, tuple, string etc. are iterables.

## Python ‘for’ Loop:
- For-loops in python corresponds to what are called **for-each loops** in many other programming languages.
- A **for** loop acts as an iterator in Python, it goes through items that are in a *collection* or any other iterable item.
    - i.e, for loop is used when we want to iterate over a collection.
    - and they request items one-by-one from a collection or more strictly from an iterable series and assign them in turn to a variable we specify.
    - for loop will continue to iterate over it's elements until there is element left in collection.
- 'for' and 'in' keywords are used.
- Collection can be iterator or iterable.
    - Examples of iterable objects are string, list, tuple, dictionary, sets, range object etc.
    - Example of iterator objects are enumerate object, map object, filter object etc.


- **Syntax:** 
```python
for var in sequence:
    block statement(s)
```

- **Actually implementation of for loop:**

```python
# create an iterator object from that iterable
iter_obj = iter(iterable)

# infinite loop
while True:
    try:
        # get the next item
        element = next(iter_obj)
        # do something with element
    except StopIteration:
        # if StopIteration is raised, break from loop
        break
```
- So internally, the for loop creates an iterator object, iter_obj by calling iter() on the iterable.
    - Ironically, this for loop is actually an infinite while loop.
    - Inside the loop, it calls next() to get the next element and executes the body of the for loop with this value. After all the items exhaust, StopIteration is raised which is internally caught and the loop ends. Note that any other kind of exception will pass through.

### Python range()
- It generates a sequence of integer values for given range.
   - It returns range object.
   - range object is iterable.

- **Syntax:** `range(start, stop, step)`
   - **start (optional):** starting number, the default is 0
   - **stop:** ending number, is not included in the generated sequence
   - **step (optional):** the difference between each item in the sequence, the default is 1
    
- **NOTE: Python range objects are not iterators. Range is a class of a list of immutable objects.**

````python
# Generate a sequence from 0 to 10 with step 1
r1 = range(11)
print(r1)
print(type(r1))
print(list(r1))
````

````python
# Generate a sequence from 5 to 10 with step 1
r2 = range(5,11)
print(r2)
print(list(r2))
````

````python
# Generate a sequence from 2 to 10 with step 2
r3 = range(2,11,2) # Only integer steps are allowed
print(r3)
print(list(r3))
````

````python
# Generate a sequence from 11 to -2 with step -1
r4= range(11,-2,-1)
print(list(r4))
````

### Using for, WAP to check whether a user input number is prime or not

````python
# Method 1:
num = int(input('Enter your number: '))
count = 0
if (num == 1):
    print('1 is neither prime nor composite')
else:
    for i in range (1, num+1):
        if (num % i == 0):
            count += 1
    if (count == 2):
        print(f'{num} is prime number')
    else:
        print(f'{num} is not a prime number')
````

````python
# Method 2:
num = int(input('Enter your number: '))
count = 0
if (num == 1):
    print('1 is neither prime nor composite')
else:
    for i in range (1, (num//2)+1):
        if (num % i == 0):
            count += 1
    if (count == 1):
        print(f'{num} is prime number')
    else:
        print(f'{num} is not a prime number')
````

````python
# Method 3:
num = int(input('Enter your number: '))
count = 0
if (num == 1):
    print('1 is neither prime nor composite')
else:
    for i in range (1, int(num**(1/2))+1):
        if (num % i == 0):
            count += 1
    if (count == 1):
        print(f'{num} is prime number')
    else:
        print(f'{num} is not a prime number')
````

### 1.
<img src='./Image/3.2 Image a.png' width=20% height=30%/>

````python
# Using for
n = int(input('Enter number of rows: '))
for i in range(1, n+1):
    for j in range(1, i+1):
        print('*', end =' ')
    print()
````

````python
# using while
n = int(input('Enter number of rows: '))
i= 1
while(i < n+1):
    j = 1
    while (j < i+1):
        print('*', end=' ')
        j += 1
    print()
    i += 1
````

### 3.
<img src='./Image/3.3 Image f.png' width=40% height=40%/>

````python
# Using for
n = int(input('Enter number of rows: '))
for i in range(n,0,-1):
    for j in range(i):
        print("*",end=' ')        
    print() 
````

````python
# using while
n = int(input('Enter number of rows: '))
i = n
while i>0:
    j = 0
    while j<i:
        print("*",end=' ')
        j += 1
    print()
    i -= 1
````

### 4.
<img src='./Image/3.2 Image b.png' width=150 height=150>

````python
# using for
n = int(input('Enter number of rows: '))
temp = n
for i in range(1, n+1):
    for j in range(1, i):
        print(" ",end =" ")
    for k in range(1,temp+1):
        print('*', end =' ')
    temp=temp-1
    print()
````

````python
# using while
n = int(input('Enter number of rows: '))
temp = n
i= 1
while (i < n+1):
    j= 1
    while (j < i):
        print(" ",end =" ")
        j = j+1
    k = 1
    while (k < temp+1):
        print('*', end =' ')
        k = k+1
    temp = temp-1
    print()
    i = i+1
````

### 5.
<img src='./Image/3.3 Image g.png' width=40% height=40%/>

````python
n = int(input('Enter number of rows: '))
for i in range(n,0,-1):
    for k in range(n-i):
        print(" ",end= ' ')   
    for j in range(i):
        print(" * ", end=' ')        
    print()
````

### 6.
<img src = './Image/3.2 Image c.png' width =40% height=30%/>

````python
ip = input("Enter a string: ")
c = 2 * ( (len(ip)//2) + 1 )
for i in range(len(ip)):
    for j in range(0, c):
        print(" ", end=" ")
    for k in range(0, i+1):
        print(ip[k], " ", end=" ")
    print()
    c=c-1
````

### 7.

````python
n = int(input())
if n%2:
    loop1 = loop2 = n
    print(' '*(n//2),'*','@'*(n+2),'*',sep='')
    for i in range(1, n):
        print(' '*(n//2),'*',' '*(n+2),'*',sep='')
    for j in range(1, n):
        for k in range(1, j):
            print(' ',end ='',sep='')
        for l in range(1, loop1+1):
            print('@',end='')
        loop1 = loop1-2
        for m in range(1, j):
            print(' ',end ='',sep='')
        for n in range(1, 4):
            print(' ',sep='',end='')
        for o in range(1, j):
            print(' ',end ='',sep='')
        for p in range(1, loop2+1):
            print('@',end='')
        loop2 = loop2-2
        print()
````

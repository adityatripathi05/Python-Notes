## Loop control Statements:

### break statement
- With the 'break' statement we can stop the loop even if the decision statement is satisfied.
- If the execution reaches a break statement, it immediately exits the loop's clause.
- In code, a break statement simply contains the break keyword.
<img src='./Image/3.2 Image f1.jpg' width=30% height=10%/>

```python
# Exit the while loop when i is 4:
i = 1
while i < 6:
    if i == 4:
        break
    print(i, end= ' ')
    i += 1
```

```python
# Exit the for loop when i is 4:
for i in range(1,7):
    if i == 4:
        break
    print(i, end= ' ')
```

### continue statement
- With the 'continue' statement we can skip the current itertion of loop even if the decision statement is satisfied.
- When the program execution reaches a continue statement, the program execution immediately jumps back to the start of the loop and re-evaluates the loop’s condition.
<img src='./Image/3.2 Image g1.jpg' width=30% height=10%/>

```python
# Continue to the next iteration inside while loop if i is 4:
i = 0
while i < 6:
    i += 1
    if i == 4:
        continue
    print(i,end=' ')    
```

```python
# Continue to the next iteration inside for loop if i is 4:
for i in range(1,7):
    if i == 4:
        continue
    print(i, end= ' ')
```

### pass statement
- The **pass statement** is a null operation, nothing happens when it executes.
- Generally used in place of block statement.

```python
# Pass
for i in 'country':
    pass
```

### WAP for Signup and login
- During signup store data only if
    - username is unique
    - length of password is greater than equal to 5.
- to restrict user to enter password upto 3 times while login

```python
import sys
acc_detail={'aditya':'123456'}
print('1. Signup','2. Login','3. Exit',sep='\n')
ch= int(input("Enter choice no.: "))
if ch==1:
    print('---------SIGN UP-----------')
    while True:
        name= input("Choose Username: ")
        if name not in acc_detail.keys():
            pwd= input("Choose Password: ")
            if len(pwd)>6:
                acc_detail[name]=pwd
                print("Account Created")
                print(acc_detail)
                break
            else:
                print("Read Guideline")
        else:
            print("Try different Username")
elif ch==2:
    print('----------LOGIN----------')
    name=input("Enter Username: ")
    if name in acc_detail.keys():
        attempt=1
        while attempt<4:
            pwd=input("Enter password: ")
            if pwd==acc_detail[name]:
                print(f'Welcome,{name}!!')
                break
            else:
                print(f"Wrong Password,{3-attempt} attempts left")
            attempt +=1
    else:
        print("Unknown User")
elif ch==3:
    sys.exit()
else:
    print("Invalid Choice")
```

### WAP to guess the correct no. between 1 to 20

```python
from random import randint
from sys import exit
secret_num=randint(1,20)
print("Think of a no. between 1 to 20.")
attempt=0
while True:
    guess= int(input("Guess the no.: "))
    if guess==0: 
        print('You loose!!')
        exit()
    elif guess==secret_num:
        print(f"Congrats you guessed correctly in {attempt} attempts.")
        break
    elif guess<secret_num:
        print("Thinking too low!!")
    elif guess>secret_num:
        print("Thinking too high!!")
    attempt+=1
```

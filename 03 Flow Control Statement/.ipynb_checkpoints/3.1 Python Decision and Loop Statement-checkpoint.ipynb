## Python Conditional Statements (Decision Making):
While programming, you will encounter many situations where your program has to execute statements based on certain conditions. These statements are called conditional statements.

The conditional statements provided by python are **‘if-elif-else’** statements.

### Python if Statement:
The ‘if’ statement in python is used to execute the statements inside the ‘if’ block if the condition satisfies. If the test expression in the ‘if’ statement is not True, then the program skips the code inside the ‘if’ block.

**Syntax:**

if(condition):

    block statement(s)
   
    
**NOTE:** space before block statement is called **indentation** and is equal to 4 spaces or 1 tab. In Python, it is required for indicating what block of code a statement belongs to.

```python
age = int(input('Enter your age: '))
if (age > 18):
    print('Eligible to vote')
```

### Python if-else Statement:
The ‘if’ statement in python is used to execute the statements inside the ‘if’ block if the condition satisfies. If the test expression in the ‘if’ statement is not true, then the statement inside ‘else’ block is executed.

**Syntax:**

if(condition):
    
    block statement(s)
    
else:
    
    block statement(s)
    
**NOTE: 'else' has no condition**

#### WAP to find whether number input by user is even or odd

```python
num = int(input('Enter a number '))
if (num % 2 == 0):
    print(f'Number {num} is even')
else:
    print(f'Number {num} is odd')
```

```python
b = str(bin(num))
print(b)
if (b[-1]=='0'):
    print(f'Number {num} is even')
else:
    print(f'Number {num} is odd')
```

### Python if-elif-else Statement:
In python, ‘if-elif-else’ allows us to have multiple ‘if’ statements i.e, to test for more than one test expression in the same ‘if-elif-else block’. The ‘elif’ stands for ‘else if’. If the test expression in the ‘if’ statement returns False, the program will check the test expression of next ‘elif’ block and so on. Python allows you to have multiple ‘elif’ statements in the same ‘if-elif-else’ block, however, there can be only one ‘else’ block.

**Syntax:**
if (condition):
    
    block statement(s)
    
elif (condition):
    
    block statement(s)

elif (condition):
    
    block statement(s)
    
else:
    
    block statement(s)

#### WAP to find greatest of three numbers

```python
num_1 = int(input('Enter 1st number: '))
num_2 = int(input('Enter 2nd number: '))
num_3 = int(input('Enter 3rd number: '))

if (num_1 > num_2 and num_1 > num_3):
    print(f'Number {num_1} is the greatest of {num_1}, {num_2}, {num_3}')
elif(num_2 > num_1 and num_2 > num_3):
    print(f'Number {num_2} is the greatest of {num_1}, {num_2}, {num_3}')
else:
    print(f'Number {num_3} is the greatest of {num_1}, {num_2}, {num_3}')
```

#### Or using multiple variable assignment

```python
num_1, num_2, num_3 = input('Enter comma seperated three numbers: ').split(',')  #split() method returns a list of strings

if (int(num_1) > int(num_2) and int(num_1) > int(num_3)):
    print(f'Number {num_1} is the greatest of {num_1}, {num_2}, {num_3}')
elif(int(num_2) > int(num_3)):
    print(f'Number {num_2} is the greatest of {num_1}, {num_2}, {num_3}')
else:
    print(f'Number {num_3} is the greatest of {num_1}, {num_2}, {num_3}')
```

## Python Loops
While programming, you will encounter many situations when you need to execute single or block of statements repeatedly. So instead of writing those statements repeatedly we use loops. 

Python supports two types of loops: **‘for’** and **‘while’**.

### Python ‘for’ Loop:
The keyword **‘in’** is used with the ‘for’ loop.

**Syntax:** 

for var in sequence:

    statement(s)

'for' loop works with sequence suchas string, list, tuple, dictionary or with range() and enumerate() but not with integers.

- **Python range():** This function generates a sequence of numbers. 

  **Syntax:** range(start, stop, step)
    - **start (optional):** starting number, the default is 0
    - **stop:** ending number, is not included in the generated sequence
    - **step (optional):** the difference between each item in the sequence, the default is 1
**NOTE: Python range objects are not iterators. Range is a class of a list of immutable objects.  
    
- **Python enumerate():** This function adds a counter to an iterable and returns it in a form of enumerate object.

   **Syntax:** enumerate(iterable, start=0)
   
   
- **Python zip():** This function is to map the similar index of multiple containers so that they can be used just using as single entity.

    **Syntax:** zip(*iterators)

```python
for i in range(1, 5):
    print(i)
```

```python
st= 'INDIA'
for i in st:
    print(i)
```

```python
for i in enumerate('INDIA', start=1):
    print(i, end='')
```

```python
players = [ "Sachin", "Sehwag", "Gambhir", "Dravid", "Raina" ]
scores = [99, 203, 58, 50, 78 ]
print('------SCORE BOARD------')
for pl, sc in zip(players, scores): 
    print(f"Player: {pl}, Score: {sc}")
```

#### WAP to check whether a user input number is prime or not

```python
# Method 1:
num = int(input('Enter your number: '))
count = 0
if (num == 1):
    print('1 is neither prime nor composite')
else:
    for i in range (1, num+1):
        if (num % i == 0):
            count += 1
    if (count == 2):
        print(f'{num} is prime number')
    else:
        print(f'{num} is not a prime number')
```

```python
# Method 2:
num = int(input('Enter your number: '))
count = 0
if (num == 1):
    print('1 is neither prime nor composite')
else:
    for i in range (1, (num//2)+1):
        if (num % i == 0):
            count += 1
    if (count == 1):
        print(f'{num} is prime number')
    else:
        print(f'{num} is not a prime number')
```

```python
# Method 3:
num = int(input('Enter your number: '))
count = 0
if (num == 1):
    print('1 is neither prime nor composite')
else:
    for i in range (1, int(num**(1/2))+1):
        if (num % i == 0):
            count += 1
    if (count == 1):
        print(f'{num} is prime number')
    else:
        print(f'{num} is not a prime number')
```

#### WAP to find all prime numbers upto user input range

```python
num = int(input('Enter your number: '))
if (num == 1):
    print('1 is neither prime nor composite')
else:
    for i in range(2, num+1):
        count = 0
        for j in range (1, i+1):
            if (i % j == 0):
                count += 1
        if (count == 2):
            print(i, end=' ')
```

### Python ‘while’ Loop
The ‘while’ loop can be used to iterate the block of statements until the condition or test expression is True. The test expression in the ‘while’ statement is checked first and if it returns True, the block inside the ‘while’ loop is executed. After all the statements inside the ‘while’ loop are executed, the control comes back to the condition in the ‘while’ loop again. If the condition still returns True, then the statements are executed again. If the test expression is False, the ‘while’ loop is terminated.

**Syntax:**

while(condition):
    
    block statement
    
    increment/decrement or any operation for loop execution
    
**Note: remember to increment/decrement, or else the loop will continue forever.**

```python
i = 1
while i < 6:
  print(i)
  i += 1
```

#### WAP to print first 'n' natural number:

```python
num = int(input('Enter the number '))
count = 1
while (count <= num):
    print(count, end =' ')
    count = count + 1
```

### Loop control Statements:

#### Break, Continue and Pass:
- With the **break statement** we can stop the loop even if the while condition is true and passes the control to the statement following immediately after loop.


- With the **continue statement** we can skip the current iteration of the loop, and continue with the next iteration.


- The **pass statement** is a null operation; nothing happens when it executes.

```python
# Exit the loop when i is 4:
i = 1
while i < 6:
  print(i)
  if i == 4:
    break
  i += 1
```

```python
# Continue to the next iteration if i is 3:
i = 0
while i < 6:
  i += 1 
  if i == 3:
    continue
  print(i)
```

```python
# Pass
for i in 'country':
    pass
```

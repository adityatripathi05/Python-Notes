## Python Loops
- While programming, you will encounter many situations when you need to execute block of code repeatedly. So instead of writing those statements again and again we use to loops over statement. 
- Python supports two types of loops: **‘for’** and **‘while’**.

## Python ‘while’ Loop
- The code in a 'while clause' will be executed as long as the 'while' statement's condition is True.
    - The 'while' statement condition is checked first and if it returns True, the block of code inside the 'while' loop is executed.
    - You can see that a while statement looks similar to an if statement. The difference is in how they behave. At the end of an 'if' clause, the program execution continues after the 'if' statement but at the end of a while clause, the program execution jumps back to the start of the while statement.
    - If the 'while' statement condition still returns True, then the statements are executed again. If the test expression is False, the 'while' loop is terminated.
- In code, a while statement always consists of the following:
    - The while keyword
    - A condition (that is, an expression that evaluates to True or False)
    - A colon
    - Starting on the next line, an indented block of code (called the while clause)

- **Syntax:**
```python
while(condition):
    block statement
    increment/decrement or any operation for loop execution
else:
    block statement
```

- **Note: remember to increment/decrement, or else the loop will continue forever.**

### WAP to print seperate list of even and odd natural number for given lower-upper range:

````python
l = int(input('Enter lower range '))
u = int(input('Enter upper range '))
even = []
odd = []
while (l <= u+1):
    if l%2==0:
        even.append(l)
    else:
        odd.append(l)
    l += 1
    
print('Even list:\n',even)
print('Odd list:\n',odd)
````

### Using while, WAP to find multiples/factors of n
- In this task you will be using a while loop to iterate over natural numbers from 1 to 100 and store only the multiples of n in a list and print that list
    - Initialize an empty list factors and a variable num equal to 1

    - Set the condtion in the while loop until the variable num is less than or equal to 100

    - Check using an if statement whether the variable is divisible by n and if is, append that number(i.e. num) to factors

    - Increment the variable num by 1 post checking the above condition. (note that this increment will be outside the if statement, take care of the indentation in order to avoid a scenario of infinite loop)

    - Print out the factors list

````python
# input value
n = int(input("Enter no.: "))

# initialize
factors=[]
num=1

# while loop
while num<100:
    if n%num==0:
        factors.append(num)
    num+=1
    
# display answer
print(factors)
````

### Using while, WAP to find HCF of two no.s

````python
a = int(input("Enter 1st number: "))
b = int(input("Enter 2nd number: "))
if (a > b):
    smaller = b
else:
    smaller = a
hcf = 1
i=1
while i<smaller+1:
    if ((a % i == 0) and (b % i == 0)):
        hcf = i
    i+=1
    
print(f'HCF of {a},{b} is {hcf}')
````

### Using while, WAP to check whether a no. is prime or not

````python
num = int(input('Enter your number: '))
count = 0
i = 1
while (i < num+1):
    if (num % i == 0):
        count += 1
    i= i+1
if (count == 2):
    print(f'{num} is prime number')
else:
    print(f'{num} is not a prime number')
````

### Using while, WAP to return a list of all prime no. between given range

````python
lower = int(input("Enter lower limit: "))
upper = int(input("Enter upper limit: "))
prime=[]
while lower < upper+1:
    c=0
    i=1
    while i < lower+1:
        if lower%i==0:
            c += 1
        i+=1
    if c==2:
        prime.append(lower)
    lower+=1
    
print(prime)
````

### Using while, WAP to check whether a 3 digit no. is Armstrong number or not.
- An Armstrong number is a number such that the sum of its digits raised to the third power is equal to the number itself.  For example, 371 is an Armstrong number, since `3**3 + 7**3 + 1**3 = 371`.

````python
n = int(input("Enter your number "))
temp = n
s=0
while temp>0:
    r = temp%10
    s += r**3
    temp = temp//10
if s==n:
    print(f'{n} is Armstrong number')
else:
    print(f'{n} is not a Armstrong number')    
````

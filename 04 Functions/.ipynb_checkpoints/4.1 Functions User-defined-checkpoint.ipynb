## Functions:
- A function is a named block of non-self executable code of instructions.
<br/><br/>
- The idea is to put some common or repeatedly done task together and make a function, so that instead of writing the same code again and again for different inputs, we can call the function.
    - For reusability of block of code or functionality, we use functions.
    - It is suggested to define different function for each individual functionality.
    - We can reuse function in same program and also in other programs by use of module.
- In python we have some builtin functions such as print(), max(), min(), count() etc and also we can create our own functions known as User defined function.


### Python Function:
- Parts of python function:
    - Header: Defined by 'def' keyword followed by name of function and a pair of parenthesis with a number of parameters separated by a comma.
    - Statements: Block of code, at an indentation level.
    - Calling the Function
<br/><br/>
- **Syntax:**

```python
def func_name(parameters):
    """
    Doc String
    """
    block of code
    return statement
    
func_name(arguments)
```

***
- **NOTE: In the syntax above:**
    - Be careful with names of function, you wouldn't want to call a function the same name as a built-in function in Python (such as len).
    - Parameters while defining function are optional.
    - We must indent to begin the code inside your function correctly. Python makes use of whitespace to organize code.
    - "doc string" is where we document(write a basic description) about the function.
    - To call a Python function at a place in our code, we simply need to name it followe by paranthesis, and pass arguments, if any.

### Docstring in Python
- Docstring is short for documentation string.
- It is a string that occurs as the first statement in a module, function, class, or method definition. We must write what a function/class does in the docstring.
- Triple quotes are used while writing docstrings.
- Docstring is available to us as the attribute `__doc__` of the function.

### Defining Function with parameter(s):
- We can define function with parameters in three types:
    - Parameters with non-default value
    - Parameters with default value
    - Variable no. of Parameters
    
 
#### Parameter with Non-Default value:
- These are those parameter which are not followed by an equal sign (=) and default value. 
- In this, we need to provide an argument while calling the function, else we get an error.

````python
def greeting(name):
    print('Hello '+name)
    
# greeting() #TypeError: 1 argument required
greeting('Aditya')
greeting('Shiva')
# greeting('Preeti','Sweety') #TypeError: Only 1 argument required
````

- These are those parameter which are followed by an equal sign (=) and default value.
- In this, if we don't provide an argument while calling the function, it's default value will be used.

````python
def greeting(name='Everyone'):
    print('Hello '+name)
    
greeting()
greeting('Aditya')
````

### Passing Arguments to Function:
- We can pass arguments to function in two ways:
    - As Positional Arguments
    - As Keyword Arguments

````python
#positional arguments
def display(name, std, marks='passing'):
    print(f'{name} got {marks} marks in class {std}.')

display('Aditya',10)
display('Aditya',10,499)
display(10,499,'Aditya') #i.e, we must pass arguments in order of definition
````

````python
#keyword arguments
def display(name, std, marks='passing'):
    print(f'{name} got {marks} marks in class {std}.')

display(name='Aditya',std=10)
display(name='Aditya',std=10,marks=499)
display(std=10,marks=499,name='Aditya') #i.e, call functions in such a way that the order of the arguments can be changed.
````

### Defining Function for Arbitrary no. Arguments:

#### Packing:
- When we do not know in advance the number of arguments that will be passed into a function.
    - `*` is used to pack arguments as tuple.
    - `**` is used to pack arguments as dictionary.

````python
def add(*args):  # the positional arguments get packed into a tuple before being passed into the function.
    print('Tuple:',args)
    s = 0
    for i in args:
        s = s+i
    print('Sum:',s)        
    
add(3,4)
add(3,4,5)
````

````python
def add(**kwargs): # the keyword arguments get packed into a dictionary before being passed into the function.
    print('Dictionary:',kwargs)
    s = 0
    for i in kwargs.values():
        s = s+i
    print('Sum:',s)
    
add(x=2, y=5)
add(x=2, y=5, z=3)
````

````python
def func(name,age,*args, **kwargs):
    print('Normal:',name,age)
    print('Tuple:',args)
    print('Dictionary:',kwargs)
    
func('Ben',30, 2,4,6, a=10, b=20)
````

#### Unpacking:
- We can use `*` or `**` to unpack the sequence or dictionary respectively so that all the elements of it can be passed as different parameters.

````python
def func(a, b, c, d):
    print(a+b+c+d)

args = [1,2,3,4]
func(*args)

kwargs = {'a':1, 'b':2, 'c':3, 'd':4}
func(**kwargs)
````

### Function with return statement
- `return` statement is optional in python and if not written then default handled by interpreter.
- `print` statement is used if we only want to display the result in python console only, but if we want to reuse the result after execution of function in the program further then use `return` statement.
- `return` statement is used to return value. This value can be a result that it produced on execution of function, or it can be something we specify- an expression or a value.
- return type of function is the type of value it return. 
    - Implicit return type of function `NoneType` and returned value will be `None`.
- As soon as a `return` statement is reached in a function, the function stops executing. Then, the next statement after the function call is executed.

````python
def add(n1, n2):
    s= n1+n2
    print('Sum is',s)
    return s
    
x = int(input("Enter 1st no.: "))
y = int(input("Enter 2nd no.: "))
res= add(x, y)
print(res, type(res))
if res%2==0:
    print('Even')
else:
    print('Odd')
````

````python
# Implicit return statement
def add(n1, n2):
    s= n1+n2
    print('Sum is',s)
    
x = int(input("Enter 1st no.: "))
y = int(input("Enter 2nd no.: "))
res= add(x, y)
print(res,type(res))
````

#### 2) Returning as List:

#### 3) Returning as Dictionary:

````python
def func(P, C, M, first, last): 
    return {'marks':P+C+M, 'name':first+" "+last}
  
op = func(60,68,70, 'Aditya', 'Tripathi')
print(op,type(op))
````

#### 4) Returning as object:
- we can create a class to hold multiple values and return an object of the class.

````python
class Store:
    def __init__(self, P,C,M, f, l):
        self.phy = P
        self.chem = C
        self.math = M 
        self.first = f
        self.last = l
        
    def addition(self):
        return self.phy+self.chem+self.math
    
    def concatenation(self):
        return self.first+" "+self.last

def func(Ph,Ch,Ma, ft, lt):
    return Store(Ph, Ch, Ma, ft, lt)

op = func(60,68,70, 'Aditya', 'Tripathi')
print(op.addition())
print(op.concatenation())
````

### Scope of a Variable:
- Scope of a variable is the portion of a program where the variable is recognized. 
- The scope for a variable depend on whether it is inside a function.
- Based on scope, there are two types of variable:
    - **Global variables** are the one that are defined and declared outside a function.
        - We can use them anywhere inside program.
    - **Local variables** are the one that are defined and declared inside a function.
        - We can use them inside function scope only.


#### Summary:
- In Python, variables are associated with two types of scopes: Local and Global
    - All the variables defined in a module contain the global scope unless or until it is defined within a function.
    - All the variables defined inside a function contain a local scope that is limited to this function itself. 
        - We can not access a local variable globally.
    - We can use global variable inside function but cannot modify it's value.
    - If two variables are defined with the same name with the two different scopes, i.e, local and global, then the priority will always be given to the local variable.
    - If global variable is defined inside function then if it's value will change inside function then will also get change outside the function definition.


### Lifetime of a Variable:
- A variable’s lifetime is the period of time for which it resides in the memory.
- The lifetime of variables inside a function is as long as the function executes. Local variables are destroyed once we return from the function.

### WAF to print all prime numbers between two numbers:

````python
def prime(lower,upper):
    for i in range (lower+1, upper):
        c = 0
        for j in range(1, i+1):
            if (i% j==0):
                c = c+1
        if(c==2):
            print(i, end = ' ')
````

````python
l = int(input("Enter lower range "))
u = int(input("Enter upper range "))
prime(l,u)
````

### WAF to print Highest Common Factor (HCF) of two numbers

````python
def computeHCF(a, b):
    if (a > b):
        smaller = b
    else:
        smaller = a
    hcf = 1
    for i in range(1, smaller+1):
        if ((a % i == 0) and (b % i == 0)):
            hcf = i
    return hcf
````

````python
num1 = int(input("Enter 1st number: "))
num2 = int(input("Enter 2nd number: "))
print(f"H.C.F of {num1} and {num2} is: {computeHCF(num1, num2)}")
````

### WAF to print fabonacci series upto n no. of terms i.e, 0 1 1 2 3 5 8....

````python
def fabo(num):
    pos1 = 0
    pos2 = 1
    if (num == 1):
        print(pos1)
    elif (num == 2):
        print(pos1, pos2)
    else:
        print(pos1, pos2, end=' ')
        for i in range(num-2):
            pos3 = pos1 + pos2
            pos1 = pos2
            pos2 = pos3
            print(pos3, end=' ')
````

````python
n= int(input("Enter no. of terms: "))
fabo(n)
````

### WAF that accepts a string and calculate no. of upper, lower case alphabets and spaces.

````python
def count(s):
    u, l, sp = 0, 0, 0
    for i in s:
        if (i.isupper() == True):
            u = u + 1
        elif (i.islower() == True):
            l = l + 1
        else:
            sp = sp + 1
    print(f'Upper case count {u}')
    print(f'Lower case count {l}')
    print(f'Space count {sp}')
````

````python
st = input('Enter your string: ')
count(st)
````

### WAF that takes in a string of more or more words and returns the same string but with those words reversed if their length is greater than 5.

````python
reversed.__doc__
````

````python
def spinWords(txt):
    words= txt.split()
    new=[]
    for w in words:
        if len(w)>5:
            new.append(reversed(w))
        else:
            new.append(w)
    txt=' '.join(new)
    return txt

msg= input("Enter your msg: ")
print(spinWords(msg))
````

## Recursion:
- In programming, recursion is when a function calls itself.
- We know that in Python, a function can call another. But when it calls itself, there must be a **base condition**, along with a **decrement statement**, to avoid an infinite loop.

### Using recursion, WAF to calculate factorial.
- The factorial of a number n is `n*(n-1)*(n-2)*..*2*1`

````python
def factR(num):
    if(num==0 or num == 1):  # Base case
        return 1
    return num*factR(num-1)  # split case
````

````python
num= int(input("Enter no.: "))
factR(num)
````

- The base case that defines to get out of the Python recursion is when n is equal to 0 or 1. 
    - In that case, the function will return 1.
- Otherwise, it will return n multiplied by factorial(n-1). This is a recursive call to itself. 
- So this is how it goes:
```python
factorial(5)
=5*factorial(4)
=5*4*factorial(3)
=5*4*3*factorial(2)
=5*4*3*2*factorial(1)
=5*4*3*2*1
=5*4*3*2
=5*4*6
=5*24
=120
```
- Process run on stack memory and it has a concept of last come first out and bottom to top approach. Hence, each digit of recursion will get printed in reverse order.

### Using recursion, WAF to print fabonacci series upto n terms

````python
# Using recursion
def faboR(num):
    if (num == 1 or num == 2):
        return num-1
    return (faboR(num-2) + faboR(num-1))
````

````python
n= int(input("Enter no. of terms: "))
for i in range(1,n+1):
    print(faboR(i), end=' ')
````

### Pass by Reference:
- In Python, every variable name is a reference. 
    - When we pass a variable to a function, a new reference to the object is created.
    
    
- Parameters to functions are references to objects, which are passed by value i.e, Object references are passed by value. 
    - When you pass a variable to a function, python passes the reference to the object to which the variable refers(the value). Not the variable itself.
    - If the value passed in a function is immutable, the function does not modify the caller’s variable. 
    - If the value is mutable, the function may modify the caller’s variable in-place.
    
<https://robertheaton.com/2014/02/09/pythons-pass-by-object-reference-as-explained-by-philip-k-dick/>

````python
def func(x,y,z):
    x = 31
    y.append(32)
    z = [33] # new reference
    print(x,'\n',y,'\n',z)

a = 11 # im-mutable object
b = [12] # mutable object
c = [13] # mutable object 
func(a,b,c)
print(a,'\n',b,'\n',c)
````

**NOTE: When we pass a reference and changes the received reference to something else then the connection between passed and received parameter is broken.** 

### WAF to guess the 6 digits lottary number.

````python
import random

def lottary_no():
    ticket_no = set() #can't initialise like {}
#     for i in range(4): 
# for loop will no give desired output as if bychance two or more random no. will be same total no. will be less than 4 as set
# don't allow repeatition
    while (len(ticket_no)<6):
        ticket_no.add(random.randint(0,9))
    return ticket_no

def gamer_no():
    guess_no = input('Enter 4 non repetitive comma seperated digits from 0-9: ').split(',')
    user_no = set()
    for i in guess_no:
        user_no.add(int(i))
    return user_no

def final_round():
    ticket_no = lottary_no()
    user_no = gamer_no()
#     matched_no = user_no.intersection(ticket_no)
#     print(f'Correctly guessed no. is/are {matched_no} and amount won is Rs.{100**len(matched_no)}')
    count = 0
    matched_no = set()
    for t, u in zip(ticket_no,user_no):
        if (t == u):
            matched_no.add(t)
            count = count + 1
    print(f'Correctly guessed no. is/are {matched_no} and amount won is Rs.{100**count}.....BINGO')   
            
final_round()
````

## Partial Function: 
Partial function allow us to fix a certain number of arguments of a function and generate a new function.

- Partial functions can be used to derive specialized functions from general functions and therefore help us to reuse our code.

````python
from functools import partial
def poly_linear(b0, b1, b2, x):  # intercept and constants of slope for polynomial L.R. equation with degree 2
    return (b0 + (b1*x) + (b2*(x**2)))

y = partial(poly_linear, 2, 0.5, 0.1)
print(y(2)) # i.e, for different values of x we will get different lines
y(5)
````

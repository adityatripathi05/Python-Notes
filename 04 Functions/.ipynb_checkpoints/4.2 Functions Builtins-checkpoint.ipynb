### WAP to take 2 array of different sizes and add even index elements of first half from 1st array to 2nd array and add odd index elements of second half from 2nd array to 1st array.

```python
ar1=[1,2,3,4,5,6]
ar2=[11,12,13,14,15,16,17,18]
car1= ar1.copy()
car2= ar2.copy()
for i in range(len(car1)):
    if i%2==0 and i<len(car1)/2:
        ar2.append(car1[i])
print(ar2)

for i,v in enumerate(car2):
    if i%2!=0 and i>len(car2)/2:
        ar1.append(v)
print(ar1)
```

### Python zip()
- This function is to map the similar index of multiple containers so that they can be used just using as single entity. 
    - i.e, it combine multiple iterables given as argument.
- It return a zip object whose next() method returns a tuple.
- Zip object is an iterator.
- **Syntax:** `zip(iterable1, iterable2,...)`

```python
# WAP to create a dictionary with certain key and value pairs stored seperately in list etc.
marks=[24,45,89,65,78]
sub=('phy','chem','maths','eng','cs')
unit=[3,6,3,3,4]
report= zip(sub,unit,marks)
print(report)
print(type(report))
print(list(report))
```

```python
marks=[24,45,89,65,78]
sub=('phy','chem','maths','eng','cs')
unit=[3,6,3,3,4]
report= zip(sub,unit,marks)
for i in report:
    print(i)
```

```python
# Converting zip object back to individual lists:
players = [ "Sachin", "Sehwag", "Gambhir", "Dravid", "Raina" ]
scores = [99, 203, 58, 50, 78 ]
board = zip(players, scores)

obj1, obj2 = zip(*board)
print(list(obj1))
print(list(obj2))
```

### Anonymous functions: lambda expression
- **lambda** keyword is used to create anonymous functions.
    - lambda expression is not declared by 'def' keyword.
    - lambda expression can have any no. of arguments but return one value for the expression we have defined it.
    - lambda expression has no name but it can be called through the variable it is assigned to.
- Usually, a function is written like: 
$f(x) = x^2 - x + 42$
- It can be written anonymously as:
$x \rightarrow x^2 - x + 42$
- **Syntax:** `lambda arguments : expression`
- **NOTE: Mostly we use lambda() within some builtin function of python like map(), reduce(), filter() etc.**

```python
# Program for addition of two no.
def func(x, y):
    return x+y
c = func(3,4)
print(c)

# Or using lambda()
func = lambda x,y : x+y
print(func(3,4))
print(func("Hello ", "Python"))
```

### map(): Executes a specified function for each item in a iterable sequence.
- **Syntax:** `map(function, iterable)`
    - function : It is a function to which map passes each element of given iterable.
    - iterable : It is a iterable which is to be mapped.
- map() return iterator map object.    
- **NOTE:** We can pass one or more iterable to the map() function.
<img src='./Image/4.2 Image a.png' width=60% height=60%/>

### Compute num^2 for each element in the list:

```python
numbers = [1, 2, 3, 4]

# Normal method:
squared = []
for num in numbers:
    squared.append(num ** 2)
print(squared)

# Using map():
def powerOfTwo(num):
    return num ** 2

squared = list(map(powerOfTwo, numbers))
print(squared)

# Using map() function and lambda expression:
squared = list(map(lambda num: num**2, numbers))
print(squared)
```

### WAF to return common values out of two list

```python
def common(l1, l2):
    l3 = []
    for i in l1:
        for j in l2:
            if ((i == j) and (i not in l3)):
                l3.append(i)
    return l3

list1 = list(map(int,input('Enter comma seperated elements: ').split(',')))
list2 = list(map(int,input('Enter comma seperated elements: ').split(',')))

common(list1, list2)
```

### reduce(): Used to apply a particular function to all of the elements of iterable.
- This function is defined in “functools” module.
- **Syntax:** `functools.reduce(function, iterable)`
<img src='./Image/4.2 Image c.png' width=60% height=60%/>
- **Working:** 
    - At first step, first two elements of sequence are picked and the result is obtained.
    - Next step is to apply the same function to the previously attained result and the number just succeeding the second element and the result is again stored.
    - This process continues till no more elements are left in the container.
    - The final returned result is returned and printed on console.

### WAP to find cumulative frequncy

```python
import functools
freq = [1, 2, 3, 4, 5]
cumulative = functools.reduce(lambda x, y: x+y, freq)
print(cumulative)
```

### classmethod()
- When we pass a method as an argument to classmethod(), it converts it into a python class method one that belongs to the class. Then, we call it like we would call any static method in python without an object.
- We can also use the syntactic sugar @classmethod for this.

```python
class LinearAlgebra:
    def simple_linear(self):
        print("y=mx+c is the equation.") 

LinearAlgebra.simple_linear= classmethod(LinearAlgebra.simple_linear)
LinearAlgebra.simple_linear()
```

```python
class LinearAlgebra:
    @classmethod
    def simple_linear(self):
        print("y=mx+c is the equation.") 
        
LinearAlgebra.simple_linear()
```

### staticmethod()
- staticmethod() creates a static method from a function. 
- A static method is bound to a class rather than to an object. But it can be called on the class or on an object.
- We can also use the syntactic sugar @staticmethod for this.

```python
class Greet:
    def sayhi():
        print("Hi")
        
Greet.sayhi=staticmethod(Greet.sayhi)
Greet.sayhi()
```

```python
class Greet:
    @staticmethod
    def sayhi():
        print("Hi")
        
Greet.sayhi()
```

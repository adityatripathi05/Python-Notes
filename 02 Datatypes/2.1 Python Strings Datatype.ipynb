## String in Python: 
- Strings in Python are ordered collection unicode characters delimited by single, double or triple quotes.
- Strings are used in Python to record text information, such as name.
- Python strings are immutable i.e, once we've constructed a string we can't modify its contents.
- Python strings are iterable objects.
<br/><br/>
- Strings in python can be indexed often synonymously called subscripted as well. 
    - In forward indexing first character of the string has index 0.
    - In backward indexing last character of string has index -1.

### Mutable vs Immutable:
- Objects which can be modified after creation in the same memory location are called mutable and the others which cannot are termed immutable.
    - Immutable Objects: int, float, long, complex, string, tuple, bool
    - Mutable Objects: list, dict, set, byte array, user-defined classes

## Access string elements
- In Python, we use brackets `[]` after an object to call its index.
- **Forward indexing** starts with 0.    
- **Backward indexing** starts with -1.


### String Indexing: 
- Used to to select the letter at a particular index/position
    - Syntax: `object[index]`

````python
print(c)
print(c[0], c[8], c[-1])
````

### Substring or Index slicing in python: 
- Process to select a subset of an entire string.
- **syntax:** ```object[start : stop : stride(step)]```
   - **start (optional):** starting number, the default is 0(in fwd indexing).
   - **stop (optional):** ending number, is not included in the generated sequence, default is len(str)(in fwd indexing).
   - **step (optional):** the difference between each item in the sequence, the default is 1(in fwd indexing).

````python
print(c)
print(c[::])
print(c[2:])
print(c[2:6])
print(c[:6])
print(c[::2])
print(c[-1::-1]) # reverse of string
````

- **String Formatting:**
    - To add formatted objects to string
        - We can use %d %f %s method
            - `%[flags][width][.precision]type` 
        - We can use the .format() method
        - We can use the f-string method

````python
name = "Einstein"
age = 22
prcent= 96.9
comp_str1= "My name is %s, my age is %d, and I secured %.2f%%." %(name, age, prcent)
print(comp_str1)
# Order and type of data is important
````

- Python's **str.format()** method of the str class allows you to carry out variable substitutions and value formatting. 
    - Formatters work by putting in one or more replacement fields or placeholders defined by a pair of curly braces {} into a string and calling the str.format() method.
    - When we pass the value we want to concatenate with the string into the method, this value will appear in the same place that your placeholder is positioned when we run the program.

````python
comp_str2= "My name is {}, my age is {}, and I secured {:.3f}%.".format(name, age, prcent)
print(comp_str2)
# formatting will be position matching
````

````python
comp_str3= "My name is {2}, my age is {0}, and I secured {1:0.3f}%.".format(age, prcent,name)
print(comp_str3)
# formatting will be through index matching
````

````python
comp_str4= "My name is {n}, my age is {a}, and I secured {per:0.3f}%.".format(a=age, per=prcent,n=name)
print(comp_str4)
# formatting will be through key matching
````

````python
comp_str5= f"My name is {name}, my age is {age}, and I secured {prcent:0.3f}%."
print(comp_str5)
# formatting will be directly with variable.
````

#### Search Functions:
- **index():** It returns the lowest index of occurrence of substring in a string. And raises ValueError if argument string is not found.
    - **Syntax:** `str.index(sub, start=0, stop=len(str))`
<br/><br/>
- **rindex():** It returns the highest index of occurrence of substring in a string. And raises ValueError if argument string is not found.
    - **Syntax:** `str.rindex(sub, start=0, stop=len(str))`

````python
print(e)
print(e.index('A'))
print(e.index('A',3))
#print(e.index('z')) # raise error
````

````python
print(e.rindex('A'))
print(e.rindex('A',3))
````

- **find():** Returns the lowest index of the substring if it is found in given string. If its is not found then it returns -1.
    - **Syntax:** `str.find(sub, start=0, stop=len(str))`
        - **sub:** It’s the substring which needs to be searched in the given string.
        - **start:** Starting position where sub is needs to be checked within the string and by default it takes 0.
        - **Stop:** Ending position where suffix is needs to be checked within the string and by default it takes length -1.
<br/><br/>
- **rfind():** Return the highest index in the string where substring sub is found.
    - - **Syntax:** `str.rfind(sub, start=0, stop=len(str))`

````python
print(e)
print(e.find('A'))
print(e.find('A',3))
print(e.find('z'))
````

````python
print(e.rfind('A'))
print(e.rfind('A',3))
print(e.rfind('z'))
````

- **count():** It counts number of times substring occurs.
    - **Syntax:** `object.count(substring, start, stop)`

````python
print(e)
print(e.count('A'))
print(e.count('A', 3))
````

- **startswith():** It is used to check whether a given string starts with some particular string.

````python
print(e)
print(e.startswith('mA'))
print(e.startswith('lE', 8))
````

- **endswith():** It is used to check whether a given Sentence ends with some particular string.

        - end index is excluded

````python
print(e)
print(e.endswith('ing'))
print(e.endswith('ne', 6))
````

#### Padding Functions
- **strip():** It removes all leading and trailing whitespace of a string and can also be used to remove particular character from leading and trailing.
    - **Syntax:** `str.strip(characters)`
    - characters (optional) : A set of characters to remove as leading characters.
<br/><br/>
- **lstrip():** It removes all leading whitespace of a string and can also be used to remove particular character from leading.
    - **Syntax:** `str.lstrip('characters')`
    - characters (optional) : A set of characters to remove as leading characters.
<br/><br/>
- **rstrip():** It removes all trailing whitespace of a string and can also be used to remove particular character from trailing.
    - **Syntax:** `str.lstrip(characters)`
    - characters (optional) : A set of characters to remove as trailing characters.

````python
str_0 = '   find me   '
print(str_0.strip())
print(str_0.lstrip())
print(str_0.rstrip())

str_1 = 'alpha'
print(str_1.strip('a'))
print(str_1.lstrip('a'))
print(str_1.rstrip('a'))
````

- **expandtabs([tabsize]):** Return a copy of the string where all tab characters are expanded using spaces. If tabsize is not given, a tab size of 8 characters is assumed.

````python
msg= "i\tlove\tmomos"
print(msg)
print(msg.expandtabs())
print(msg.expandtabs(2))
````

- **centre():** It return a centered string of given width.
    - **Syntax:** `str.center(width, fillchar=' ')`
    - Padding is done using the specified fill character (default is a space).

````python
str_6 = 'python'
print(str_6.center(8))
print(str_6.center(10, '|'))
print(str_6.center(15, '|'))
````

#### String Substitution Functions
- **replace():** It returns copy of string where all occurrences of 'old substring' have been replaced by 'new substring'.
    - **Syntax:** ```str.replace(old, new, ?)```
    - where ? is count whose default value is -1 i.e, means replace all occurrences.

````python
print(e)
print(e.replace('n','N'))
print(e.replace('n','N',1))
````

- **split():** Splits a string according to delimiter(space if not provided) and returns a list of substrings, split into at most num substrings if given. If maxsplit is given, at most maxsplit splits are done, the leftmost ones.
    - Syntax: `str.split(delimiter=" ", maxsplit=str.count(delimiter))`
<br/><br/>
- **rsplit():** Return a list of the words in the string, using sep as the delimiter string. If maxsplit is given, at most maxsplit splits are done, the rightmost ones.
    - Syntax: `str.rsplit(delimiter=" ", maxsplit=str.count(delimiter))`

````python
txt= 'Python is funny language'
print(txt.split())
print(txt.split(' ',2))
print(txt.rsplit(' ',2))
````

- **splitlines([keepends]):** Return a list of the lines in the string, breaking at line boundaries.

````python
para='''Hello Everyone. 
Welcome to Python class. 
How are you?'''
print(para.splitlines())
````

## Bytes Datatypes:
- Working of an interconversion of different data types to bytes(), usually useful for encoding schemes.
- byte() constructor converts an object to immutable byte represented object of given size and data.
- Bytes accept string as argument and requires an encoding scheme with it to perform it.
- **Syntax:** `bytes([source[, encoding[, errors]]])`
    - where,
        - source (Optional) - source to initialize the array of bytes.
        - encoding (Optional) - if source is a string, the encoding of the string.
        - errors (Optional) - if source is a string, the action to take when the encoding conversion fails
- The bytes() method returns a bytes object which is an immmutable (cannot be modified) sequence of integers in the range 0 <=x < 256.
- Bytes literals prefixed with a lowercase b.
- Convert str to bytes with encode(), bytes to str with decode().

````python
string = "Python is interesting."
# string with encoding 'utf-8'
arr = bytes(string, 'utf-8')
print(arr)
````

- bytes is a sequence of bytes, str is a sequence of Unicode codepoints

````python
# Create a byte of given integer size
size = 5
arr = bytes(size)
print(arr)
````

````python
# Convert iterable list to bytes
rList = [1, 2, 3, 4, 5]
arr = bytes(rList)
print(arr)
````

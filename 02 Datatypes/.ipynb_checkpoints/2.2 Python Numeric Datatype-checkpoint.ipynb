## Numeric  in Python:
- Integers, floating point numbers and complex no. comes under numeric datatype in python. They are defined as int, float and complex class in python.
- **Integers** are positive or negative whole numbers. 
    - Python integers are signed and have unlimited precision.
    - For examaple: 2 and -4 are integers.
- **Floats** are any real number or fractional numbers or numbers containing a decimal point(.) or a number defined using an exponential(e).
    - Python floats are implemented using IEEE-754 double precision floating-point numbers with 53 bits of binary precision.
        - This is equivalent to between 15 to 16 significant digits in decimal.
    - For example: 2.3, -2.0, 4E2 (4 times 10 to the power of 2)
    - Note that numbers written in a scientific notation such as 2.5*10^5 is a float number because it contains a decimal point in the notation.

### Type Conversion
- During data manipulation there might arise a need for converting an integer to float, float to an integer, float to a string, or splitting on a decimal point in a float.
- Let's say you have number x. The following are some common operations:
    - **int(x)** converts it to an integer.
        - i.e, Using int() constructor we can convert from other numeric types, suchas float to integers
            - float value is rounded to largest integer smaller than the given value.
        - Can also define base while converting str to integer using int(), where default base is 10.
    - **float(x)** converts it to a float.
    - **complex(x)** converts it to a complex number with real part x and imaginary part 0.

```python
print(int(3.5)) # positive float to integer with base 10
print(int(-3.5)) # negative float to integer with base 10
print(int('497')) # string to integer with base 10
print(int("10000",3)) # string to integer with base 3
```

```python
print(float(7)) # integer to float
print(float('nan')) # string nan to not a number float
print(float('inf')) # string inf to positive infinity
print(float('-inf')) # string inf to negative infinity
```

```python
# Dynamic numeric input
d = int(input("Enter a integer: "))
print(d, type(d))

e = float(input("Enter a number: "))
print(e, type(e))
```

### Python Arithmetic Operator:
- Arithmetic operators in Python are used to perform arithmetic operations such as addition, subtraction, multiplication, division etc.
<br/><br/>
    - 1. **+ (Addition):** Adds the operands. **Eg:** x + y
<br/><br/>
    - 2. **– (Subtraction):** Performs subtraction of right operand from the left operand. **Eg:** x – y
<br/><br/>
    - 3. ***(Multiplication):** Multiplies two operands. **Eg:** x * y
<br/><br/>
    - 4. **/ (Division):** Divides the operand at the left side by the operand at the right side. **Eg:** x / y
<br/><br/>
    - 5. **% (Modulus):** Returns the remainder of a division of the left operand by the right operand. **Eg:** x % y
<br/><br/>
    - 6. **// (Floor division):** Returns the largest integer smaller than the quotient of division operation. **Eg:** x // y
<br/><br/>
    - 7. ****(Exponentiation):** Returns the left-hand operand power right-hand operand. **Eg:** x ** y
<br/><br/>
    - 8. **abs(x):** Return absolute value of x.
<br/><br/>
    - 9. **exp(x):** Return e raised to the power x. e(Euler's No.) is a mathematical constant that is the base of natural logarithm i.r, 2.71828
<br/><br/>
    - 10. **log(x):**	Return logarithm with base e.
<br/><br/>
    - 11. **log10(x):** Logarithm with base 10.
<br/><br/>
    - 12. **pow(x,y):** x raised to the power y.

```python
num_1, num_2= 20, 3
```

```python
print(f'Addition of {num_1} and {num_2} is {num_1 + num_2}')

print(f'Subtraction of {num_1} and {num_2} is {num_1 - num_2}')

print(f'Multiplication of {num_1} and {num_2} is {num_1 * num_2}')

print(f'Division of {num_1} and {num_2} is {num_1 / num_2}')

print(f'Floor Division of {num_1} and {num_2} is {num_1 // num_2}')

print(f'Floor Division of {-num_1} and {num_2} is {-num_1 // num_2}')

print(f'Modulus of {num_1} and {num_2} is {num_1 % num_2}')

print(f'Exponent of {num_1} and {num_2} is {num_1 ** num_2}')

print(f'Exponent of {num_1} and {num_2} is {pow(num_1,num_2)}')

print(f'Absolute of {-num_1} is {abs(-num_1)}')
```

### Python Assignment Operators:
- In python, the operators that are used to assign values to variables are called assignment operators. The major assignment operator is an equal sign (=). 
<br/><br/>
    - 1. **=:** Assigns value at the right side to the variable at left. **Eg:** a = 100, assigns 100 to variable a.
<br/><br/>
    - 2. **+=:** Adds the operand at the right side to the variable at left and assigns the result to the variable at left. **Eg:** a += 1, equivalent to a = a + 1.
<br/><br/>
    - 3. **-=:** Subtracts the operand at the right side from the variable at left and assigns the result to the variable at left. **Eg:** a -= 1, equivalent to a = a – 1.
<br/><br/>
    - 4. ***=:** Multiplies the operand at the right side to the variable at left and assigns the result to the variable at left. **Eg:** a *= 2, equivalent to a = a * 2.
<br/><br/>
    - 5. **/=:** Divides the variable at the left side by the operand at right and assigns the result to the variable at left. **Eg:** a /= 2, equivalent to a = a / 2.
<br/><br/>
    - 6. **%=:** Performs modulo operation on two operands and assigns the result to the variable at left. **Eg:** a %= 2, equivalent to a = a % 2.
<br/><br/>
    - 7. ****=:** Performs an exponential operation on the operands and assigns the result to the variable at left. **Eg:** a ***= 2, equivalent to a = a ** 2.
<br/><br/>
    - 8. **//=:** Performs floor division on the operands and assigns the resulting value to the left variable. **Eg:** a //= 2, equivalent to a = a // 2.

```python
a = int(input("Enter no. "))
b = int(input("Enter no. "))
```

```python
a += b
print(f"Addition is {a}")
```

```python
a -= b
print(f"Subtraction is {a}")
```

```python
a *= b
print(f"Multiplication is {a}")
```

```python
a /= b
print(f"Division is {a}")
```

```python
a //= b
print(f"Floor division is {a}")
```

```python
a %= b
print(f"Remainder is {a}")
```

```python
a **= b
print(f"Power is {a}")
```

### Python Comparison (Relational) Operators:
- Also called Relational operators, they compare the values on either side of the operand and decide the relation among them.
    - Return True or False based on the relation.
<br/><br/>
- The comparison operators allowed in python are:
    - 1. **== (Equal to):** Returns True if the operands at both sides are equal, returns False otherwise. **Eg:** a == b.
<br/><br/>
    - 2. **!= (Not equal to):** Returns True if the operands are not equal, returns False otherwise. **Eg:** a != b.
<br/><br/>
    - 3. **> (Greater than):** Returns True if the left operand is greater than the operand at right, returns False otherwise. **Eg:** a > b.
<br/><br/>
    - 4. **>= (Greater than or equal to):** Returns True if the left operand is either greater than or equal to the operand at right, returns False otherwise. **Eg:** a >= b.
<br/><br/>
    - 5. **< (Less than):** Returns True if the left operand is less than the operand at right, returns False otherwise. **Eg:** a < b.
<br/><br/>
    - 6. **<= (Less than or equal to):** Returns True if the left operand less than or equal to the operand at right, returns False otherwise. **Eg:** a <= b.

```python
num_1, num_2= 20, 3

print(f'Is {num_1} equal to {num_2}: {num_1 == num_2}')

print(f'Is {num_1} not equal to {num_2}: {num_1 != num_2}')

print(f'Is {num_1} greater than {num_2}: {num_1 > num_2}')

print(f'Is {num_1} greater than or equal to {num_2}: {num_1 >= num_2}')

print(f'Is {num_1} less than {num_2}: {num_1 < num_2}')

print(f'Is {-num_1} less than or equal to {num_2}: {-num_1 <= num_2}')
```

### Python Logical Operators:
- The logical operators in python allow us to combine two or more conditional operators and return True or False based on the results of those conditional operators combined with logical operators. 
<br/><br/>
- The logical operators in python are:
    - 1. **and:** Returns True if both the conditions are True otherwise returns False. **Eg:** (a > 10) and (a < 20).
<br/><br/>
    - 2. **or:** Returns True if one of the conditions is True, otherwise returns False. **Eg:** (a > 10) or (a < 20).
<br/><br/>
    - 3. **not:** Also called negation, this operator is used to reverse the logical state of its operand. **Eg:** not (a > 10).

```python
num1, num2 = 20, 3
print(f'{(num1>15)} logical_and {(num2<5)} is {(num1>15) and (num2<5)}')
print(f'{(num1<15)} logical_and {(num2<5)} is {(num1<15) and (num2<5)}')

print(f'{(num1>15)} logical_or {(num2<5)} is {(num1>15) or (num2<5)}')
print(f'{(num1<15)} logical_or {(num2<5)} is {(num1<15) or (num2<5)}')

print(f'logical_not of {num1>15} is {not(num1>15)}')
print(f'logical_not of {num1<15} is {not(num1<15)}')
```

## Precision Handling in Python:
Handle precision of floating point numbers:
- **1) Using %:** % operator is used to format as well as set precision in python.


- **2) Using format():** This is yet another way to format the string for setting precision.


- **3) Using round(x,n):** This function takes 2 arguments, number and the number till which we want decimal part rounded.

```python
import math
pi = math.pi
print('pi:', pi)

var = round(pi, 4)#roundoff after 4 palce of decimal
print('pi:',var)

print('pi: %.3f' %pi)#roundoff after 3 palce of decimal

print('pi: {0} roundoff after 2 of place decimal= {0:.2f}'.format(pi))#roundoff after 2 of place
```

### Using math module:
- Ref: <https://docs.python.org/3/library/math.html>
<br/><br/>
- **4) math.trunc():** This function is used to eliminate all decimal part of the floating point number and return the integer without the decimal part.
<br/><br/>
- **5) math.ceil():** This function is used to print the least integer greater than the given number.
<br/><br/>
- **6) math.floor():** This function is used to print the greatest integer smaller than the given integer.

```python
import math
print('Trunc value of pi is', math.trunc(pi))

print('Ceil value of pi is', math.ceil(pi))

print('Floor value of pi is', math.floor(pi))
```

```python
# Other math module functions:
print('Pi power 2 is', math.pow(pi, 2))

print('Square root of pi is', math.sqrt(pi))

print('Factorial of 5 is', math.factorial(5))

print('Absolute value of -5 is', math.fabs(-5))

print('Remainder when of 5/3 is', math.fmod(5,3))

print('Greatest Common Divisor of 8,12 is', math.gcd(8,12))

print('Exponential of 0 is', math.exp(0))

print('Log of 1 is', math.log(1))

print('sin value of 2 radian is', math.sin(2)) # x in radian
```

### Python bit functions on int:
- Integral abstract base class, the int type implements the numbers.
    - **1. int.bit_length():** Returns the number of bits required to represent an integer in binary, excluding the sign and leading zeros.
    - **2. int.to_bytes(length, byteorder, *, signed=False):** Return an array of bytes representing an integer.
    - **3. int.from_bytes(bytes, byteorder, *, signed=False):** Returns the integer represented by the given array of bytes.

```python
n1 = 7
n2 = -7
print(n1.bit_length()) 
print(n2.bit_length()) 
```

```python
n3 = 1024
print(n3.to_bytes(2, byteorder ='big'))  # big endian machine
```

```python
print(int.from_bytes(b'\x04\x00', byteorder ='big')) 
```

### Calculate the compound interest (C.I.)
- Formula for simple interest for amount P after n years is
    - `I= (P*R*T)/100`
- Formula for compound interest for amount P after n years is
    - `A = P(1 + r/100)^n`
    - Here, A is the Amount after n years, r is the rate of interest, and P is the principal.
- Task is to calculate the amount after two years on a principal amount of USD 1000 at the rate of 10 % p.a.

```python
# store variable names
P,r,n = 1000, 10, 2

# compound interest formula
A = P*pow((1+(r/100)),n)

# display compound interest
print(A)

# display interest
interest= A-P
print(interest)
```

## Bool Datatype:
- Boolean type represents logical state.
    - There are two bool values: `True` `False` (Both initial Capitals)
- Plays an important role in python control flow structure.
    - Python decision statements and while loops accepts bool values into condition
- bool() constructor can be used to convert other types to bool.
    - For integers 0 is consider False and all other values are considered True.
    - For float also 0.0 is considered False and all other real values are considered True
    - For collection literals, empty collection is considered False.
- Bool values are commonly produced by python relational operator used for comparing objects.

```python
print(bool(0), bool(0.0))
```

```python
print(bool(100), bool(-1), bool(53.4))
```

```python
print(bool([])) # empty list
```

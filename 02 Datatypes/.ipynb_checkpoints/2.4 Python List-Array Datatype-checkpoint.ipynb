## List in python:
- List in python aare ordered collection of items of different data types.  
- The items within the list are separated by commas (,) and delimited by square brackets `[]`.
- List support forward and backward indexing.
- List are mutable.
- List are iterable objects.

### Create empty list:

```python
list1 = []
list2 = list()
print(list1, type(list1), len(list1))
print(list2, type(list2), len(list2))
```

### Create non-empty list:

```python
# Creating a list and Store student details: Roll,name,board,percentage,(subject)
stud_1= [101,'Aditya','CBSE',95.6,('Phy','Chem')]
print(stud_1, type(stud_1), len(stud_1))
```

```python
# User defined list:
l = []
n = int(input("Enter no. of elements: "))
for i in range(n):
    l.append(eval(input()))
print(l)
```

### Convert Different datatypes into list:

```python
name=list('ADITYA')
print(name, type(name))
```

```python
marks=list((67,56,89,54,78))
print(marks, type(marks))
```

## Accessing elements of list:
- In Python, we use brackets `[]` after an object to call its index.
- **Forward indexing** starts with 0.
- **Backward indexing** starts with -1.

### List Indexing:

```python
# Access roll no.
print(stud_1[0])
```

```python
# Access percentage
print(stud_1[-2])
```

```python
# Access last character of name
print(stud_1[1][-1])
```

### List Index Slicing:

```python
# Access complete list using index slicing
print(stud_1[::])
```

```python
print(stud_1[-1::-1])  # reverse the list
```

## Python Operators in list:

- **Assignment Operator(=):** To update an element or a range of items.

```python
# Updating single element
print(stud_1)
print(stud_1[-1])
stud_1[-1] = 'Hello'
print(stud_1)
```

```python
# Updating multiple elements
print(stud_1)
print(stud_1[-2:-4:-1])
stud_1[-2:-4:-1] = [21,'Hello']
print(stud_1)
```

- **Concatenation Operator(+):**
    - Process to combine two list.

```python
data = ['Phy','Chem']
new_stud = stud_1 + data
print(new_stud)
```

- **Replication Operator(+):**
    - Process of repeating elements of list multiple times.

```python
print(data*2)
```

- **Membership Operator(in/not in):** 

```python
print(95.6 in stud_1)
print('Bye' in stud_1)
print('Bye' not in stud_1)
```

### Python List methods
- **copy():** It returns a copy of given object. A copy can be classified into two types:
    - **Deep Copy:** A copy of object is copied in another object.
        - Means any changes made to a copy of object don't reflect in original object.
    - **Shallow Copy:** A reference of object is copied in another object. 
        - Means that any changes made to a copy of object do reflect in the original object.

```python
print(new_stud)
temp_stud = new_stud # Will create new reference of object
temp_stud[0]='Bee'
print(temp_stud)
print(new_stud) # Original also get changed
```

```python
print(new_stud)
stud = new_stud.copy() # Will create copy of object
stud[0]='India'
print(stud)
print(new_stud) # Original remain unchanged
```

- **reverse():** Reverse the elements of the list in place.

```python
print(stud_1)
stud_1.reverse()
print(stud_1)
print(stud_1[-1::-1])
```

- **count():** Return the number of times an element appears in the list.
    - **Syntax:** `object.count(element)`

```python
print(stud_1)
print(stud_1.count('Hello'))
```

- **index():** It returns the lowest index of occurrence of element in list. And raises ValueError if element is not found.
    - **Syntax:** `object.index(element, start, stop)`

```python
print(stud_1)
print(stud_1.index('Hello'))
print(stud_1.index('Hello', 1))
```

```python
# print(stud_1.index('Bye'))
```

- **append():** Adds its argument as a **single object** at the end of a list.
    - **Syntax:** `object.append(element)`
    - **NOTE: append() takes exactly one argument.**

```python
print(stud_1)
stud_1.append(67.8)
print(stud_1)
```

- **extend():** Iterate over its argument(s) and **add each object** at the end of  list.
    - **Syntax:** `object.extend(sequence)`
    - **NOTE: extend() takes exactly one argument.**

```python
print(stud_1)
stud_1.extend('END')
print(stud_1)
```

```python
print(stud_1)
stud_1.extend([21, 21])
print(stud_1)
```

- **insert():** It inserts the object at a **given index** in list.
    - **Syntax:** `object.insert(index, element)`

```python
print(stud_1)
stud_1.insert(3, 'Python')
print(stud_1)
```

- **remove():** Removes the first occurence of element from list. Raise ValueError if element not in list. 
    - **Syntax:** `list.remove(element)`
    - **NOTE: remove() takes exactly one argument.**

```python
print(stud_1)
stud_1.remove(21)
print(stud_1)
```

```python
# print(stud_1)
# stud_1.remove('Bye')
# print(stud_1)
```

- **pop():** It removes the object at **specified index** or **by default pop last object**. Also return the popped object.
    - **Syntax:** `list.pop(index=-1)`
    - **NOTE: pop() takes at most 1 argument**

```python
print(stud_1)
stud_1.pop()
print(stud_1)
```

```python
print(stud_1)
temp = stud_1.pop(2)
print(temp)
print(stud_1)
```

- **clear():** It removes all the elements from the list.
    - **Syntax**: `lis.clear()`

```python
stud_3= stud_1.copy()
print(stud_3)
stud_3.clear()
print(stud_3)
```

- **del():** Delete a Python list, some of its elements, or a single element.
    - **Syntax:** `del list[start:stop:step]`

```python
#Deleting a few elements: Use the slicing operator in python to delete a slice.
stud_4= stud_1.copy()
print(stud_4)
del stud_4[3:6]
print(stud_4)
```

```python
#Deleting a single element: Delete a single element use its index.
print(stud_4)
del stud_4[2]
print(stud_4)
```

```python
# Remove list object from memory
print(stud_4)
del stud_4
# print(stud_4)
```

### WAP to check whether two strings are anagram:
- 'Python is great' 
- 'Thorn pestiagy'
- These two strings are anagram of each other

```python
x=input("Enter String 1: ")
y=input("Enter String 2: ")

x= x.lower().replace(' ','').replace('',',')[1:-1].split(',')
y= y.lower().replace(' ','').replace('',',')[1:-1].split(',')
x.sort()
y.sort()
if len(x)==len(y):
    if x==y:
        print("Anagram")
    else:
        print("Not Anagram")
else:
    print("Not Anagram")
```

- **sort():** Sort the items of the list in place.
    - **Syntax:** `list.sort(key=None, reverse=False)`

```python
arr2= arr.copy()
print(arr2)
arr2.sort() # by default this will sort the array in ascending order
print(arr2)
```

```python
arr3= arr.copy()
print(arr3)
arr3.sort(reverse=True) # this will sort the given list in descending order
print(arr3)
```

```python
fruits= ['banana', 'apple', 'kiwi', 'banana', 'pear', 'apple', 'orange', 'grape']
fruits.sort()
print(fruits)
```

```python
# Custom sorting based on length of name
fruits.sort(key=len)
print(fruits)
```

## Array module: 
- Array is not standard data type in python like string, integer etc. We have to import array module if we want to use it.
- **Syntax:** `array.array(typecode,[initializers])`
    - **Where typecode:**
        - i - signed integer of size 2 bytes
        - f - float of size 4 bytes
        - d - floating point of size 8 bytes
        - c - character of size 1 byte

```python
import array
```

```python
ar1 = array.array('i',[1,2,3,4,5])
print(ar1)
print(type(ar1))
```

```python
print(ar1[1])
```

```python
print(ar1[2:5])
```

```python
ar1.append(6)
print(ar1)
```

```python
ar1.extend([9,3])
print(ar1)
```

```python
ar1.insert(3,1)
print(ar1)
```

```python
ar1.remove(3)
print(ar1)
```

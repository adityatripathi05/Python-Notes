## Python bit functions on int:
The int type implements the numbers.Integral abstract base class.

- **1) int.bit_length():** Returns the number of bits required to represent an integer in binary, excluding the sign and leading zeros.


- **2) int.to_bytes(length, byteorder, *, signed=False):** Return an array of bytes representing an integer.


- **3) int.from_bytes(bytes, byteorder, *, signed=False):** Returns the integer represented by the given array of bytes.

```python
n1 = 7
n2 = -7
print(n1.bit_length()) 
print(n2.bit_length()) 
```

```python
n3 = 1024
print(n3.to_bytes(2, byteorder ='big'))  # big endian machine
```

```python
print(int.from_bytes(b'\x04\x00', byteorder ='big')) 
```

### List Comprehension:

**WAP Create a list of squares from 1, 10**

```python
cu = []
for i in range(1, 11):
    cu.append(i**3)
    
print(cu)
```

```python
cu = [i**3 for i in range(1,11)]
print(cu)
```

**WAP to print all even no. between 1 to 20**

```python
pm = [i for i in range(2, 20) if (i%2 == 0)]
pm
```

**WAP to print all even no. between 1 to 20 with positive sign and all odd with negative sign`**

```python
e_o = [i if (i%2 == 0) else -i for i in range(1,20)]
e_o
```

#### WAP for Given an integer, n, and n space-separated integers as input, create a tuple, t, of those n integers. Then compute and print the result of hash(t) .

Sample Input:

2

1 2

Sample Output:

3713081631934410656

```python
n = int(input('Enter no. of variable to input: '))
l = []
counter = 0
while (counter < n-1):
    num = input('Enter your ssv: ').split()
    for i in num:
        l.append(int(i))
    counter +=1
t = tuple(l)
print(hash(t))
```

```python
if __name__ == '__main__':
    n = int(input())
    counter = 0
    while (counter < n-1):
        integer_list = map(int, input().split())
        counter +=1
    t = tuple(integer_list)
    print(hash(t))
```

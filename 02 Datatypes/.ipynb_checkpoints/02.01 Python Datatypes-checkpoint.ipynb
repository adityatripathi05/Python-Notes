## Numeric in Python:
Integers, floating point numbers and complex no. comes under numeric datatype in python. They are defined as int, float and complex class in python.

```python
a, b, c = 2, 2.5, 2.5+2j
```

- **type()** is used to check which class a variable or a value belongs to.

```python
print(f'{a} is of type {type(a)}')
print(f'{b} is of type {type(b)}')
print(f'{c} is of type {type(c)}')
```

- **isinstance()** is used to check if an object belong to a particular class. Return True/false

**Syntax:**

isinstance(var, type)

```python
print(f'Result of division is always of float type : {isinstance((b/a), float)}')
```

## String in Python: 
The python strings are the sequences of unicode characters within single or double quotes. Triple quotes (”’, "””) can be used for multi-line strings.

Strings in python can be indexed often synonymously called subscripted as well. First character of the string has index 0.

#### Create empty string:

```python
s = ''
print(type(s))
```

#### Create non-empty string:

```python
a, b ,c = 'bye', 'Python3', '12321' 
x, y = 'MacHine LearNinG', "Deep Learning"
z = '''Machine Learning is the subset of the Artificial Intelligence'''
```

#### Access string elements: 
Square brackets can be used to access elements of the string.

**Forward indexing** starts with 0 upto the (len(string)-1).
    
**Backward indexing** starts with -1 upto negative of len(string). More precisely since last character of string is stored as null character hence on backward indexing we negect 0 index.

```python
x[6]
```

```python
x[-7]
```

#### Substring or Index slicing in python:
syntax: string[start : stop : step]
   - **start (optional):** starting number, the default is 0
   - **stop:** ending number, is not included in the generated sequence
   - **step (optional):** the difference between each item in the sequence, the default is 1

```python
x[8:13]
```

```python
x[8:]
```

```python
x[:8:2]
```

```python
y[-1::-1]
```

```python
a*2  # String replication
```

```python
'M' in x  # Membership operator: in and not-in
```

```python
'y' not in x
```

```python
len(x)
```

**find():** The find() method returns the lowest index of the substring if it is found in given string. If its is not found then it returns -1.

**Syntax:** str.find(sub,start,end)
   - **sub:** It’s the substring which needs to be searched in the given string.
   - **start:** Starting position where sub is needs to be checked within the string and by default it takes 0.
   - **end:** Ending position where suffix is needs to be checked within the string and by default it takes length-1.

```python
print(z.find('the'))
z.find('the', 25)
```

```python
x.find('is')
```

## List in python:
- It is an ordered sequence of items. 
- The items in the single list can be of different data types. 
- The items in the list are separated by commas (,) and enclosed within square brackets [ ].

#### Create empty list:

```python
list1 = []
list2 = list()
print(type(list1))
print(type(list2))
```

#### Create non-empty list:

```python
list_1= [2, 5.4, 'hi', 600, 'bye']
```

#### Accessing elements of list:
Square brackets can be used to access elements of the list.

**Forward indexing** starts with 0 upto the (len(list)-1).
    
**Backward indexing** starts with -1 upto negative of len(list).

```python
list_1[2]
```

```python
list_1[-3]
```

```python
list_1[2][1]
```

**Assignment in list:**

```python
list_1[-1] = 'Hello'
print(list_1)
```

Hence list is mutable.

**Operations on list:**

```python
len(list_1)
```

```python
list_2 = [9, 12, 21, 5, 8]
```

```python
max(list_2)
```

```python
min(list_2)
```

```python
list_2.sort() # by default this will sort the given list in ascending order.
print(list_2)
```

```python
list_2.sort(reverse = True)   # this will sort the given list in descending order.
print(list_2)
```

```python
list_3= [12, 32, 21, 31]
```

```python
list_3.reverse()
print(list_3)
print(list_3[-1::-1])
```

**append():** Adds its argument as a **single object** to the end of a list.

**NOTE: append() takes exactly one argument.**

```python
list_3.append(['hi', 64])
print(list_3)
```

```python
list_3[-1][0][1]
```

**extend():** Iterate over its argument(s) and **add each object** to the list and extend the list.

```python
list_3.extend([21, 21, 21])
print(list_3)
```

**insert():** It inserts the object at a **given index** in a list.

**Syntax:** list.insert(index, object)

```python
list_3.insert(3, 'Python')
print(list_3)
```

```python
list_3.count(64)
```

**index():** Return index of object.

**Syntax:** list.index(object, start, stop)

```python
print(list_3.index(21))
list_3.index(21, 6)
```

```python
list_4= list_2 + list_3  # Concatenate list
print(list_4)
```

```python
list_3 * 2  # replicate list
```

**del():** This method deletes all the elements in range.

**Syntax:** del list[start : end]

```python
del list_4[1: 5]
print(list_4)
```

```python
l = [2,2,5,'ok','hi']
l.remove(2)
print(l)
```

**remove():** Removes particular element from list

**Syntax:** list.remove(element)

**NOTE:** remove() takes exactly one argument.

```python
list_4.remove(['hi', 64])
print(list_4)
```

**pop():** It removes the object at **specified index** or **by default pop last object**. Also return the popped object.

**NOTE:** pop() takes at most 1 argument

```python
x= list_4.pop(2)
print(x)
print(list_4)
```

## Tuple in python:
- Tuples are ordered sequence of items.
- The items in the single tuple can be of different data types.
- The items in the tuples are separated by commas (,) and enclosed within parentheses(Parentheses is optional).

**NOTE: Python tuples are also similar to lists except that tuples are immutable i.e, the items cannot be modified once the tuple is created.**

**Hence append, insert, remove, del like operations can't be performed in tuple.**

#### Create an empty tuple:

```python
t0 = ()
t1 = tuple()
print(type(t0))
print(type(t1))
```

#### Create non-empty tuple:

```python
tup_1 = (21, 2, 'ok', [23, 'hello'])
```

**Access elements of tuple:**

```python
tup_1[2]
```

```python
tup_1[3][1]
```

```python
tup_1[-1]
```

```python
tup_1[1:]
```

#### Hence tuple indexing and slicing are same as list.

#### Creating a tuple with single element:

```python
tup_2 = (23,)
```

#### Operations on Tuple:

```python
tup_3= tup_1 + tup_2  # Tuple concatenation
print(tup_3)
```

```python
tup_5= tup_3 * 2      # Tuple replication
tup_5
```

```python
tup_5.index(2, 4)
```

```python
tup_5.count(2)
```

```python
tup_4 = (23, 43, 15, 7, 9)
```

```python
len(tup_4)
```

```python
min(tup_4)
```

```python
max(tup_4)
```

```python
del(tup_4)
```

## Python Dictionary:
- Dictionary allows you to store in key-value pairs. 
- The **keys in the dictionaries must be unique** as they are used to access the values associated with them.

**NOTE:** If key of two values are same then the last Value stored get access on calling value through key since python is interpreted language.
- The items are enclosed within braces { }. The key and value in each key-value pair are separated by a colon (:) with the key at the left side and value at the right side of the colon. The key-value pairs in the dictionaries are separated by commas (,).
- keys must be of immutable datatype suh as numeric, string or tuple and values can be of any data type supported by python. 
- Dictionary is mutable.
- Dictionary is iterable.
- Dictionary doesn't support indexing and slicing.
- After python version 3.5, dictionary are ordered pair of collection.

#### Create empty dictionary:

```python
d0 = {}
d1 = dict()
```

```python
print(type(d0))
print(type(d1))
```

#### Create non-empty dictionary:

```python
data1 = {'name':'aman', 21:'age', 'num':[50, 65, 67]}
```

```python
len(data1)
```

**Get all keys:**

```python
print(data1.keys())
list(data1.keys())
```

**Get all values:**

```python
print(data1.values())
list(data1.values())
```

**Get all key-value:**

```python
print(data1.items())
list(data1.items())
```

**Add item to dictionary:**

```python
data1['sub'] = ['C', 'C++']
data1
```

**Update dictionary:**

```python
data1['name'] = 'vishal'
print(data1)
```

```python
dict0 = {'name':'adi', 'age':21, 'roll':200}
dict1 = {'name':'priya', 'age':21, 'roll':201, 'sub':['Jave', 'Python']}
dict0.update(dict1)  # updating dict0
print(dict1)
print(dict0)
```

**Copy dictionary:**

```python
data2 = data1.copy() # Shallow copy
print(data2)
```

```python
data2['name'] = 'rohit'
print(data2)
print(data1)
```

**NOTE:** **Deep copy:**

```python
data3= data2
data3['name'] = 'adarsh'
print(data3)
print(data2)
```

```python
len(data2)
```

**Return value of specific key:**

```python
data2.get(21)
```

```python
data2[21]
```

**NOTE: If key is not available then get() will return None by default or will print user defined message whereas values called by indexing method will return error if key is not found in dictionary**

#### WAP to create user defined dictionary:

```python
d2 = {}
r = int(input("Enter no. of pairs to strore: "))
for i in range(r):
    k = input('Enter Key: ')
    v = input('Enter Value: ')
    d2[k] = v
print(d2)
```

#### WAP to create a dictionary with certain key and value pairs stored seperately in list etc.

```python
k = {1,2,3,4}
v = ['a', 'ad', 'adi', 'adit']
d = dict(zip(k, v))
print(d)
```

#### Program to store username and password using dictionary:

```python
details = {}
u_name  = input('Create your username: ')
pwd = input('Create your password: ')
details[u_name] = pwd
```

```python
name = input('Enter your username: ')
if (name in details.keys()):
    pas = input('Enter your password: ')
    if (pas in details.values()):
        print('Welcome')
    else:
        print('Wrong password')
else:
    print('Unknown user')
```

#### WAP to restrict user to enter password upto 3 times:

```python
details = {}
u_name  = input('Create your username: ')
pwd = input('Create your password: ')
details[u_name] = pwd
```

```python
name = input('Enter your username: ')
if (name in details.keys()):
    attempt = 3
    current_attempt = 1
    while (current_attempt < attempt+1):
        pas = input('Enter your password: ')
        if (pas in details.values()):
            print(f'Welcome {name}')
            break
        else:
            print('Wrong password')
            if(current_attempt==3):
                print('Account Locked')
                break
            print(f'Attempts left {3-current_attempt}')

        current_attempt = current_attempt+1
            
else:
    print('Unknown user')
```

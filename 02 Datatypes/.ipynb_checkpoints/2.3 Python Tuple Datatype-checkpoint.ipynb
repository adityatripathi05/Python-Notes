### Data Structure:
- Data Structure is a data organising, management and storage format that enables efficient access and modification.
- Existing datatypes suchas int, float, str, bool have some problems suchas:
    - Data in a variable can be stored in a single format.
    - Large memory overhead by creating large no. of variables.
    - Unfit for storing large structured data i.e, absence of inheritance structure in variable.
- Data Structures in python to solve these problems:
    - Tuple
    - List
    - Dictionary
    - Sets
    - Array

## Tuple in python:
- Tuples in python are ordered collection of different data types.
- The items in the tuples are separated by commas (,) and delimited by parentheses(Parentheses is optional).
- Tuples are immutable.
- Tuples are iterable objects.
<br/><br/>
- **NOTE: Python tuples are also similar to lists except that tuples are immutable i.e, the items cannot be modified once the tuple is created.**
- **Hence append(), extend(), insert(), remove(), del like operations can't be performed in tuple.**

### Create an empty tuple:

```python
t0 = ()
t1 = tuple()
print(type(t0), len(t0))
print(type(t1), len(t1))
```

### Create non-empty tuple:

```python
# Creating a tuple with single element: Store Grade
grade = ('A',)
```

```python
# Creating a tuple and Store student details: Roll,name,board,percentage
stud_1 = 101,'Aditya','CBSE',95.6
stud_1 = (101,'Aditya','CBSE',95.6)
stud_1 = tuple((101,'Aditya','CBSE',95.6))
print(stud_1, type(stud_1), len(stud_1))
```

```python
# User defined tuple:
t = []
ip = input("Enter csv for list: ").split(',')
for i in ip:
    if i.isalpha():
        t.append(i)
    elif i.isdigit():
        t.append(int(i))
    else:
        t.append(float(i))
print(tuple(t))
```

### Access elements of tuple

```python
# Access roll no.
print(stud_1[0])
```

```python
# Access percentage
print(stud_1[-1])
```

```python
# Access last character of name
print(stud_1[1][-1])
```

```python
# Access complete tuple using index slicing
print(stud_1[::])
```

```python
# Access name,board,percentage
print(stud_1[1:])
```

```python
# Access only roll,board
print(stud_1[::2])
```

```python
# Reverse the details of student
print(stud_1[::-1])
```

### Python Operators in Tuple:

- **Assignment Operator(+):**

```python
# Roll of student was stored incorrectly, correct it with 102 value
# stud_1[0]= 102 # Error: tuple' object does not support item assignment
print(stud_1)
```

- **Concatenation Operator(+):**

```python
# Join grade with other details
detail= stud_1 + grade
print(detail)
```

- **Replication Operator(*):**

```python
multi_grade= grade * 2      # Tuple replication
print(multi_grade)
```

### Python Tuple Methods:

```python
# Check the index of Grade
print(detail.index('A'))
```

```python
# check how many times percentage is stored
print(detail.count(95.6))
```

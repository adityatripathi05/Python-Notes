## Python Sets:
- Sets in Python are unordered collection of different datatype(immutable).
    - Sets have no duplicate elements.
- The items in set are separated by commas (,) and enclosed within curly braces {}.
- Sets are mutable.
- Sets are iterable objects.

- Types of sets:
    - **Set type** is mutable -- the contents can be changed using methods like add() and remove().
    - **Frozen type** sets are immutable objects that only support methods and operators that produce a result without affecting the frozen set or sets to which they are applied. Hence it can be used as a dictionary key or as an element of another set.

### Create Empty Sets

```python
s0 = set()
print(type(s0), len(s0))
```

**NOTE: An empty set can't be declared by only clurly braces eg. set1 = {}, as it is declaration of empty dictionary.**

### Create a Non-Empty Sets

```python
s0 = {3, 4.5, 'hi', 7}
```

```python
# User defined set:
s1 = []
ip = input("Enter csv of set: ").split(',')
for i in ip:
    if i.isalpha():
        s1.append(i)
    elif i.isdigit():
        s1.append(int(i))
    else:
        s1.append(float(i))
print(set(s1))
```

```python
# print(s0[1]) # give error since 'set' object is not subscriptable
```

Hence neither indexing, slicing nor assignment is possible in sets

```python
print(3 in s0)     # membership operation
print('Hey' in s0) 
```

```python
s2 = {3, 'hi', 5, 12, 'hey'}
```

```python
print(s0.union(s2))  # new set with elements from both s0 and s2
print(s0|s2)
```

```python
print(s0.intersection(s2))  # new set with elements common to s0 and s2
print(s0 & s2) 
```

```python
print(s0.difference(s2))  # new set with elements in s0 but not in s2
print(s0 - s2)
```

```python
print(s0.symmetric_difference(s2))  # new set with elements in either s0 or s1 but not both
print(s0 ^ s2)
```

```python
print(s0.copy())  # new set with a shallow copy of s0
```

```python
s3 = {'hi', 7}
print(s3.issubset(s0))   # test whether every element in s3 is in s0
print(s3 <= s0)
```

```python
print(s0.issuperset(s3))   # test whether every element in s2 is in s0
print(s0 >= s3)
```

```python
s0.add('Hello')   # Adds the item x to set if it is not already present in the set.
print(s0)
```

```python
s4 = {101, 'Python'}   # return set s0 with elements added from s4
s0.update(s4)
print(s0)
```

```python
s0.remove('Hello')    # remove x from set s0; raises KeyError if not present
print(s0)
```

```python
s0.discard('hi')    # removes x from set s0 if present
print(s0)
```

```python
s0.pop()   # remove and return an arbitrary element from s0
print(s0)
```

```python
s0.clear()   # remove all elements from set s0
```

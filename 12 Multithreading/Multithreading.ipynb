## Multitasking:
- Multitasking is a facility to execute multiple jobs concurrently(by using the concept of time sharing).


### Advantages of Multitasking:
- a) It improves the efficiency of application.
- b) Here the jobs run independently without affecting others jobs.


### Types of Multitasking:
- 1) Process Based Multitasking (Multiprocessing)
- 2) Thread based Multitasking (Multithreading)


### Process and Multiprocessing:
- Program in execution is called process. Ex.
    - Running a browser is a process
    - Running a video game is a process
<br/><br/> 
- **Any process has 3 basic components:**
    - An executable program.
    - The associated data needed by the program (variables, work space, buffers, etc.)
    - The execution context of the program (State of process)    
<br/><br/>
- For each process resources are allocated independently.
<br/><br/>
- Concurrent execution of multiple process is called **multiprocessing**.
- Multiprocessing is OS level concept.
- Ex. of multiprocessing is simultaneously operating vlc,ms-word, bowser.

    
### Threads and Multithreading:
- A thread is the smallest unit of execution with the independent set of instructions.
- Thread is part of the process and operates in the same context sharing program’s runnable resources like memory.
    - The thread exist entirely inside process and share it's(process) resources.
- A thread has 
    - a starting point, 
    - an execution sequence 
    - a result
    - has an instruction pointer that holds the current state of the thread and controls what executes next in what order.
- Every process has at least one thread, which is the process itself, i.e, main thread. 
    - This main thread actually creates the child thread objects.
<br/><br/> 
- Ex. While playing a video game, the game as a whole is a single process and system has to handle multiple tasks 
    - handle graphics
    - handle audio
    - handle inputs for user
    - handle user interface
    - handle networking (when playing multiplayer)
- All these tasks are performed simultaneously in parallel and also remaning responsive all the times. For this system we use concept of multithreading.
<br/><br/>
- Inside a process multiple threads can work together to accomplish a common goal. Concurrent execution of multiple threads is called **multithreading**.
    - i.e, threads allows the program to run tasks in parallel. The operating system executes these threads like parallel "processes". On a single processor machine, this parallelism is achieved by thread scheduling or timeslicing.

- Multithreading is a program level concept.
- Ex. multiple tabs opened in browser is example of multithreading.

### Types of Thread:
- **1. Kernel thread:** These threads are the part of operating system.
- **2. User thread:** User space threads are not implemented in kernel. The user space thread can be seen as an extension of function concepts in programming languages. 
    - i.e, user-space thread is similar to a function call. But there are differences to regular functions, especially the return behaviour.


### Inside a Thread:
- A thread contains all this information in a **Thread Control Block (TCB)**:
    - <img src= './Image/12.1 Image c.png' width=45% height=35%/>
    - **Thread Identifier:** Unique id (TID) is assigned to every new thread
    - **Stack pointer:** Points to thread’s stack in the process. Stack contains the local variables under thread’s scope.
    - **Program counter:** a register which stores the address of the instruction currently being executed by thread.
    - **Thread state:** can be running, ready, waiting, start or done.
    - **Thread’s register set:** registers assigned to thread for computations.
    - **Parent process Pointer:** A pointer to the Process control block (PCB) of the process that the thread lives on.


### Advantages of multithreading:
- 1. Multithreaded programs can run faster on computer systems with multiple CPU's because theses threads can be executed truly concurrent.
- 2. A program can remain responsive to input. This is true both on single and multiple CPU's.
- 3. Allows to do something else while one thread is waiting for Input/Output task to complete.
- 4. Some programs are easy to express using concurrency.
- 5. Threads of a process can share the memory of global variables. If a global variable is changed in one thread, this change is valid for all threads. A thread can have local variables.

    - other one is file handling.
    - If working without implementation of multithreading, when 1st job is done only then 2nd job will be done. 
        - Total time of execution of process will be total time to complete all the task of program sequentially. 
        - And if in case in any one job there comes exception during execution whole program will get terminated without any job completion.
    - But if we see clearly one job require user interference for input but the other job don't require user intervention i.e, both jobs are independent of each other. Hence we can apply multithreading.
        - Here if both jobs are done using multithreading boths jobs can be done effectively and independently.
        - Now if in case in any one job there comes exception during execution whole program will get not terminated and other threaded process will get completed.

#### Implement multithreading

```python
from time import time_ns
from threading import Thread, current_thread, activeCount
def even_odd(num):
    print('Check for even-odd',time_ns())
    if(num % 2 == 0):
        print(num,"is Even")
    else:
        print(num,"is Odd")
    print('Check for even-odd Done',time_ns())
        
def file_handling():
    print('Open a file to write',time_ns())
    f1 = open('File2Save/msg.txt', 'w')
    f1.write('Hello\n')
    print('Writing done',time_ns())
    f1.close()

    print('Open a file to read',time_ns())
    f2 = open('File2Save/msg.txt')
    data= f2.read()
    print('Reading done',time_ns())
    f2.close()

    print(data)
    print('File handling done',time_ns())

if __name__=='__main__':
    print('Main thread start',time_ns())
    num=int(input("Enter a no.: "))
    t1=Thread(target=even_odd, args=(num,))
    t2=Thread(target=file_handling)

    print('Even-odd thread triggered',time_ns())
    t1.start()
    print('File-Handling thread triggered',time_ns())
    t2.start()
    
    print('Total thread in program:', activeCount())
    print('Main thread completed',time_ns())
```

- **To create a new thread, we create an object of Thread() class. It takes following arguments:**
    - **target:** the function to be executed by thread
        - if target is function with parameters then argument is given in args as tuple.
        - t1= Thread() creates thread in memory but will not execute untill target argument is passed.
    - **args:** the arguments to be passed to the target function
<br/><br/>
- In above example, we created 2 threads with different target functions.
<br/><br/>
- **To start a thread, we use start() method of Thread class.**
    - start() creates stack memory for that thread and loads the job

```python
from time import time_ns
from threading import Thread, current_thread, activeCount
def even_odd(num):
    print('Check for even-odd',time_ns(), current_thread().getName())
    if(num % 2 == 0):
        print(num,"is Even")
    else:
        print(num,"is Odd")
    print('Check for even-odd Done',time_ns(), current_thread().getName())
        
def file_handling():
    print('Open a file to write',time_ns(), current_thread().getName())
    f1 = open('File2Save/msg.txt', 'w')
    f1.write('Hello\n')
    print('Writing done',time_ns(), current_thread().getName())
    f1.close()

    print('Open a file to read',time_ns(), current_thread().getName())
    f2 = open('File2Save/msg.txt')
    data= f2.read()
    print('Reading done',time_ns(), current_thread().getName())
    f2.close()

    print(data)
    print('File handling done',time_ns(), current_thread().getName())

if __name__=='__main__':
    print('Main thread start',time_ns(), current_thread().getName())
    num=int(input("Enter a no.: "))
    t1=Thread(target=even_odd, args=(num,), name='Th1')
    t2=Thread(target=file_handling, name='Th2')

    print('Even-odd thread triggered',time_ns(), current_thread().getName())
    t1.start()
    print('File-Handling thread triggered',time_ns(), current_thread().getName())
    t2.start()
    
    print('Total thread in program:', activeCount())
    print('Main thread completed',time_ns(), current_thread().getName())
```

- Once the threads start, the current program (you can think of it like a main thread) also keeps on executing. 
- In order to stop execution of current program until a thread is complete, we use **join()** method.
    - As a result, the current program will first wait for the completion of other started threads.
    - Once, they are finished, the remaining statements of current program are executed.

```python
from time import time_ns
from threading import Thread, current_thread, activeCount
def even_odd(num):
    print('Check for even-odd',time_ns(), current_thread().getName())
    if(num % 2 == 0):
        print(num,"is Even")
    else:
        print(num,"is Odd")
    print('Check for even-odd Done',time_ns(), current_thread().getName())
        
def file_handling():
    print('Open a file to write',time_ns(), current_thread().getName())
    f1 = open('File2Save/msg.txt', 'w')
    f1.write('Hello\n')
    print('Writing done',time_ns(), current_thread().getName())
    f1.close()

    print('Open a file to read',time_ns(), current_thread().getName())
    f2 = open('File2Save/msg.txt')
    data= f2.read()
    print('Reading done',time_ns(), current_thread().getName())
    f2.close()

    print(data)
    print('File handling done',time_ns(), current_thread().getName())

if __name__=='__main__':
    print('Main thread start',time_ns(), current_thread().getName())
    num=int(input("Enter a no.: "))
    t1=Thread(target=even_odd, args=(num,), name='Th1')
    t2=Thread(target=file_handling, name='Th2')

    print('Even-odd thread triggered',time_ns(), current_thread().getName())
    t1.start()
    print('File-Handling thread triggered',time_ns(), current_thread().getName())
    t2.start()
    
    t1.join() # blocks the execution of current execution thread untill calling thread terminates or timeout.
    t2.join()
    
    print('Main thread completed')
```

### Multithreading by extending the Thread class:
- When a child class is created by extending the Thread class, the child class represents that a new thread is executing some task. 
- When extending the Thread class, the child class can override only two methods i.e. the `__init__()` method and the run() method. 
    - No other method can be overridden other than these two methods.

```python
#Python multithreading example to print current date.
#1. Define a subclass using threading.Thread class.
#2. Instantiate the subclass and trigger the thread.

from threading import Thread, current_thread
import datetime

def print_date(threadName, counter):
    datefields = []
    today = datetime.date.today()
    datefields.append(today)
    print(f"{threadName}[{counter}]: {datefields[0]}")
    
class myThread (Thread):
    def __init__(self, name, counter):
        Thread.__init__(self)
        self.name = name
        self.counter = counter
        
    def run(self):
        print("\nStarting " + self.name)
        print_date(self.name, self.counter)
        print("Exiting " + self.name)

# Create new threads
thread1 = myThread("Thread", 1)
thread2 = myThread("Thread", 2)

# Start new Threads
thread1.start()
thread1.join()

thread2.start()
thread2.join()

print("\nExiting the Program!!!")
```

### Issues related with multithreading:
- **1. Scheduling:** To execute a threaded program it must repeatedly switch between threads. So a program should be able to switch between threads efficiently and conveniently.
<br/><br/>
- **2. Resource sharing:** Since threads of a process can share the memory of global variables and other resources so operation performed in one thread could cause problem in another thread.
<br/><br/>
- **3. Synchronization:** Since threads often need to coordinate their actions hence the outcomes depened on the order of thread execution. And often for synchronization we use locking primitives.

### Race Conditions:
Race conditions can occur when two or more threads access a shared piece of data or resource.

#### Thread locking and sequentation:
If multiple threads are doing their job by common resource, sometimes we need to provide this common resource in mutually exclusive manner. To implement this concept we use locking.

- In programming we need to create object of 'Lock' class. i.e, lock = Lock()


- A thread can acquire lock by calling acquire() method on the lock object. i.e, lock.acquire()


- A thread that has already acquire the lock must release this lock by calling release() method on the same lock object. i.e, lock.release()

```python
from threading import *
import time

lock = Lock()

def print_table(num):
    lock.acquire()
    for i in range(1,11):
        print(num * i, end= ' ')
        time.sleep(1)
    lock.release()
    print('\ncompleted\n')

t1 = Thread(target = print_table, args = (5, ))
t2 = Thread(target = print_table, args = (6, ))

t1.start()
t2.start()
```

#### join():
This method blocks the execution of current execution thread untill calling thread terminates or timeout.

```python
from threading import *
import time

lock = Lock()

def print_table(num):
    lock.acquire()
    for i in range(1,11):
        print(num * i, end= ' ')
        time.sleep(1)
    lock.release()
    print('\nThread completed\n')

t1 = Thread(target = print_table, args = (5, ))
t2 = Thread(target = print_table, args = (6, ))

t1.start()
t2.start()

t1.join()
print('Main thread completed') # Since it is part of main thread
```

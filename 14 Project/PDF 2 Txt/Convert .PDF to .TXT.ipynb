```python
import PyPDF2
```

```python
with open('sample.pdf','rb') as pdfobj:
    pdf_reader = PyPDF2.PdfFileReader(pdfobj)
    n = pdf_reader.numPages
    pgobj = pdf_reader.getPage(n-1)
    txt_data = pgobj.extractText()
```

```python
with open('sample.txt','w+') as txtobj:
    txtobj.writelines(txt_data)
```

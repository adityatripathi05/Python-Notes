### Speech Recognition
- Speech recognition is the ability of a computer software to identify words and phrases in spoken language and convert them to human readable text.
- Here we'll learning how to use 'SpeechRecognition' Python library for performing speech recognition to convert audio speech to text in Python.
- This library supports several recognition engines:
    - Google Speech Recognition
    - Google Cloud Speech API
    - IBM Speech To Text
    - Microsoft Bing Voice Recognition
    - Wit.ai
    - Houndify API
    - CMU Sphinx (offline)
    - Snowboy Hotword Detection (offline)
- Google Speech Recognition doesn't require any API key.
- Install the library: `pip install SpeechRecognition`

```python
import speech_recognition as sr
# initialize the recognizer
r = sr.Recognizer()
```

```python
sr.__version__
```

## Transcribe a WAV audio file

```python
# Sample Test
with sr.AudioFile("long_speech.wav") as source:     # use "long_speech.wav" as the audio source
    audio = r.record(source)                        # read and extract audio data from the file
try:
    # for testing purposes, we're just using the default API key
    # to use another API key, use `r.recognize_google(audio, key="GOOGLE_SPEECH_RECOGNITION_API_KEY")`
    # instead of `r.recognize_google(audio)`
    print("Transcription: " + r.recognize_google(audio))   # recognize speech using Google Speech Recognition
except sr.UnknownValueError:                                # speech is unintelligible
    print("Google Speech Recognition could not understand audio")
except sr.RequestError as e:
    print("Could not request results from Google Speech Recognition service; {0}".format(e))
```

```python
# Function to reading from the audio file of .wav format
def audiofile2text(audio):
    with sr.AudioFile(audio) as source:
        audio_data = r.record(source)
        try:
            return "Transcription: " + r.recognize_google(audio_data)
        except sr.UnknownValueError:
            print("Google Speech Recognition could not understand audio")
        except sr.RequestError as e:
            print("Could not request results from Google Speech Recognition service; {0}".format(e))
```

```python
# specify filename/path
file= "long_speech.wav"

# call audiofile2text()
output= audiofile2text(file)
print(output)
```

### Code for converting the microphone speech into text:
- NOTE: this requires PyAudio because it uses the Microphone class
- `install the library: conda install -c anaconda pyaudio`

```python
# Test Sample
import speech_recognition as sr
r = sr.Recognizer()
with sr.Microphone() as source:                # use the default microphone as the audio source
    print("Start Speaking..5 Sec")
    audio = r.record(source, duration=5)       # record for the first phrase and extract it into audio data
    print("Stop")
try:
    print("You said:" + r.recognize_google(audio))    # recognize speech using Google Speech Recognition
except sr.UnknownValueError:
    print("Google Speech Recognition could not understand audio")
except sr.RequestError as e:
    print("Could not request results from Google Speech Recognition service; {0}".format(e))
    
# write audio to a WAV file
with open("microphone-results.wav", "wb") as f:
    f.write(audio.get_wav_data())
```

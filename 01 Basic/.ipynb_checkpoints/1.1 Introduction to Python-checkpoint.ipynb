## Static Vs Dynamic Programming Language:
- **Static programming language** are those languages in which we can't write a code for performing logical operations using variables etc. Eg HTML etc.
- **Dynamic programming language** are those languages in which we can write a code for performing logical operations using variables etc. Eg. C, C++, Python, Java etc.

## Statically typed Vs Dynamically typed Programming Language:
- **Static typed programming languages** are those in which 
    - 1. variables need not be defined before they’re used.
        - Static typing does not imply that you have to declare all the variables first, before you use them; variables maybe be initialized anywhere, but developers have to do so before they use those variables anywhere.
    - 2. variable types are explicitly declared and thus are determined at compile time. Ex. C, C++, Java etc.
    - 3. type checking (i.e. the process of verifying and enforcing the constraints of types) is done at compile-time(a static check).
        - <img src='./Images/1.1 Image f.png' width=20% height=20%/>
        - This lets the compiler decide whether a given variable can perform the actions requested from it or not.
    - 4. In Statically typed languages, once a variable has been declared with a type, it cannot ever be assigned to some other variable of different type and doing so will raise a type error at compile-time.
<br/><br/>
- **Features of Static typed programming languages:**
    - A large class of errors are caught in the early stage of development process.
    - Static typing usually results in compiled code that executes more quickly because when the compiler knows the exact data types that are in use, it can produce optimized machine code (i.e. faster and/or using less memory).

- **Dynamic typed programming languages** are those languages in which 
    - 1. variables must necessarily be defined before they are used. 
    - 2. it do not require the explicit declaration of the data type of variables. Ex. Python, PHP etc
    - 3. type checking is done at run-time(dynamic check).
        - <img src='./Images/1.1 Image g.png' width=10% height=10%/>
    - 4. In Dynamically typed languages, variables are bound to objects at run-time by means of assignment statements, and it is possible to bind the same variables to objects of different types during the execution of the program.


- **Features of Dynamic typed programming languages:**
    - Dynamic type checking typically results in less optimized code than static type checking. 
    - It also includes the possibility of run time type errors and forces run time checks to occur for every execution of the program (instead of just at compile-time).
    - Implementations of dynamically type-checked languages generally associate each run time object with a type tag (i.e., a reference to a type) containing its type information. This run-time type information (RTTI) can also be used to implement dynamic dispatch, late binding, down-casting, reflection, and similar features.
    - The absence of a separate compilation step means that you don’t have to wait for the compiler to finish before you can test your code changes. This makes the debug cycle much shorter and less cumbersome.

## Strong Typing Vs Weak Typing:
- **Strong Typing:** Programming languages in which variables have specific data types are **strong typed**. 
    - This implies that in strong typed languages, variables are necessarily bound to a particular data type.
    - Ex. Python, Java.
<br/><br/>    
- **Weak Typing:** Weak typed languages are those in which variables are not of a specific data type. 
    - It should be noted that this does not imply that variables do not have types; it does mean that variables are not "bound" to a specific data type. 
    - Ex. PHP and C.

### Summary: 
- Python is dynamic typed i.e, there is no type checking of the code prior to running it.
    - and strong typed i.e, there is no way to circumvent that type
- Java is static typed and strong typed. 
- PHP is dynamic typed and weak typed. 
- C is static typed and weak typed.

### Python is Interpreted
Many languages are compiled, meaning the source code you create needs to be translated into machine code, the language of your computer’s processor, before it can be run. Programs written in an interpreted language are passed straight to an interpreter that runs them directly.

This makes for a quicker development cycle because you just type in your code and run it, without the intermediate compilation step.

One potential downside to interpreted languages is execution speed. Programs that are compiled into the native language of the computer processor tend to run more quickly than interpreted programs. For some applications that are particularly computationally intensive, like graphics processing or intense number crunching, this can be limiting.

## Architecture of Python:
<img src='./Images/1.1 Image d.png' width=60% height=50%/>

- **Source file** contains high level instructions written by programmer. Extension of this file is `.py`.
<br/><br/>
- **Python Compiler** is the integral part of python interpreter and checks the syntax by compiling the program. 
    - If there is syntax error, the output of the program will be "SyntaxError".
    - If there is no syntax error it will generate a intermediate code i.e, `byte code` from source file. Extension of this file is `.pyc`.
    - Compiled code is run by python interpreter.
<br/><br/>        
- **Bytecode** is program code that has been compiled from source code into low-level code designed for a software interpreter. Byte code converted to machine code everytime when the program runs.
    - Bytecode is an implementation detail of the CPython interpreter.
    - .pyc files Python leaves lying around aren't just some "faster" or "optimized" version of our source code; they're the bytecode instructions that will be executed by Python's virtual machine as our program runs.
    - Prevents Python from having to re-parse the source code every time the program runs.
    - The byte-code is loaded into the Python run-time and interpreted by a virtual machine, which is a piece of code that reads each instruction in the byte-code and executes whatever operation is indicated. 
<br/><br/>
- **PVM (Python Virtual Machine)** is the integral part of python interpreter and it generate low level code form python byte code in "on the fly mode" i.e, saved no where. Machine code is directly handover to CPU.
    - Steps in PVM:
        - **Step 1) Laxing:** It is responsible for dividing program into tokens. (Tokens are smallest unit of a program)
        - **Step 2) Parsing:** Tree(Bunch of nodes) formation on which token is placed at each node.
        Parent node and child node
        - **Step 3) Compiler:** Convert high level language to low level language. In python convert unicode code to binary code.
        - **Step 4) Execution:** Operation is performed and output is generated in binary form which is again converted to unicode code and final output on output screen.
<br/><br/>
- **NOTE: After completing the execution of program interpreter delete the .pyc file. Machine code is not saved in interpreter but we can save it by using it as a module**

## Python 2.x Vs Python 3.x:

- **1) Division Operator:**
    * In Python 2.x, the return type of a division operation (/) depends on its operands. We need to use the floating value i.e, 7.0/5 and -7/5.0 as operand to get desired output of 1.4 and -1.4 respectively. Otherwise the result will be 1 and -2 respectively i.e, round the calculation to the nearest whole number.
    * In Python 3.x, division (/) operation always returns a float.
<br/><br/>
- **2) print Statement:**
    * In Python 2.x, a print pair of parentheses is not mandatory i.e, can be used as a property also.
    * In Python 3.x, it is mandatory to use parentheses with print i.e, print()
<br/><br/>
- **3) Encoding format:**
    * Python 2.x, has implicit ASCII string type, a separate unicode type, but there is no byte type.
    * In Python 3.x, the implicit str type is Unicode and and byte type.
<br/><br/>
- **4) xrange():**
    * Python 2.x, has both range and xrange functions. xrange is generally faster & saves memory. range() returns a list i.e, range(3) returns [0, 1, 2] while xrange() returns a xrange object i.e, xrange(3) returns an iterator object.
    * In Python 3.x, the range() now does what xrange() does in Python 2.x, so to keep our code portable.
<br/><br/>
- **5) Error Handling:**
    - In Python 3.x, `'as'` keyword is required.
<br/><br/>
- **6) Input Function:**
    - In Python 2.x, when using input(), python automatically converts the data type based on the input. And raw_input() gets the input as text.
    - In Python 3.x, the input function acts like raw_input() from Python 2.x and it always returns string type.
<br/><br/>
- **7) Round Function:**
    - In Python 2.x, round() always results in a floating point number.
    - In Python 3.x, round() return results in n digit precision.

### Python Interactive Console : Python interpreter or Python shell
- The Python interactive console (Python shell) provides programmers with a quick way to execute commands and try out or test code without creating a file. The interactive console offers the opportunity to explore Python
- Python Shell provide access to all of Python's built-in functions and standard module as well as to any installed modules.
- The python command-line environment is a **Read-Evaluate-Print-Loop(REPL)** i.e, 
    - Interpreter will read whatever we'll type in,
    - Evaluate it,
    - Print the result,
    - and then loop back to the beginning i.e, Read again
- Within interactive python session, we can enter fragments of Python programs and see instant results.
    - Within REPL We can use the special underscore variable to refer the most recently printed value. (this trick only works with REPL)
- We are currently using the Jupyter notebook, which uses an advanced Python interpreter called IPython.
- This gives us much more power and flexibility.
- Exit REPL and get back to System Shell prompt

- Features of IDLE:
    - IDLE provides support for Windows/Linux/Mac OS X/All Tk Platforms

    - multi-window text editor with multiple undo, Python colorizing, smart indent, call tips, auto completion, and other features
    - search within any window, replace within editor windows, and search through multiple files (grep)

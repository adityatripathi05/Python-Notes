# Language Primitives
- **Lexical analysis:** A Python program is read by a parser. Input to the parser is a stream of tokens, generated by the lexical analyzer i.e, lexical analyzer breaks a file into tokens.

## Tokens:
- Token is the smallest unit inside the given program.
- Python breaks each logical line into a sequence of elementary lexical components known as tokens.
- Each token corresponds to a substring of the logical line.
- The normal token types are identifiers, keywords, operators, delimiters, and literals.

## Python Keywords
- Python keywords are the reserved words in python and have a specific meaning, intended to perform some action.
    - There are as many as 33 such keywords in Python, each serving a different purpose. Together, they build the vocabulary of the Python language.
- Keywords are **case sensitive**. 
    - Mostly Keywords contain lowercase letters only except True, False, None.
    - Some keywords begin simple statements or clauses of compound statements, while other keywords are operators.
- They represent the syntax and structure of a Python program. We cannot use keywords as regular identifiers.

- **‘keywords’** is a Python builtin module.
    - For now, a module is simply a python script having a collection of related attributes. We’ll learn about modules in detail later in this tutorial.
- **‘kwlist’** is one of the attributes of the keyword module which returns a list of all Python Keywords.
- **iskeyword()** is an function of keyword module. It test whether a keyword is valid or not.
<br/><br/>
- **print()** is a builtins module function. Prints the values to a output by default.
- **len()** is a builtins module function. It returns the length of a sequence.

- Ref.: [Keywords](https://docs.python.org/3/reference/lexical_analysis.html#keywords)

````python
import keyword
kw= keyword.kwlist
print(kw)
print(f'\nTotal no. of keywords in python: {len(kw)}')
````

````python
print(keyword.iskeyword("func"))
print(keyword.iskeyword("try"))
````

## Python Identifiers
- Python Identifiers are user-defined names to represent a variable, function, class, module or any other object.. 
- It helps in differentiating one entity from another.
<br/><br/>
- **Python Identifier writing rules:**
    - An identifier starts with a letter (A to Z or a to z) or an underscore (_) followed by zero or more letters, underscores, and digits (0 to 9). 
        - For example, myvar, student_Class, employee0, _var1, var_100 are valid identifiers.
    - An identifier cannot be started with a digit character. 
        - For example, myvar9 is valid identifier while 9myvar is not.
    - Case is significant in Python: lowercase and uppercase letters are distinct.
    - Python keywords are not allowed to be used as identifiers.
    - Python does not allow punctuation characters such as '@', '$', and '%' within identifiers.
    - Normal Python style is to start class names with an uppercase letter and all other identifiers with a lowercase letter.
    - Starting an identifier with a single leading underscore indicates by convention that the identifier is meant to be private. 
    - Starting an identifier with two leading underscores indicates a strongly private identifier; if the identifier also ends with two trailing underscores, the identifier is a language-defined special name.
<br/><br/>  
- **isidentifier():** Test whether a Python identifier is valid or not.

````python
print('_var1'.isidentifier())
print('9func'.isidentifier())
````

## Python Literals
- Literals can be defined as a data that is given in a variable or constant.
- Python support the following literals:
    - **1. String literals:** A string literal is a sequence of characters surrounded by quotes. Eg. 'xyz' or "xyz"
        - We can use both single, double or triple quotes for a string. 
        - And, a character literal is a single character surrounded by single or double quotes.
        <br/><br/>
    - **2. Numeric literals:** Numeric Literals are immutable(unchangeable). Eg. 2, 2.5, 2+5j
        - Numeric literals can belong to 3 different numerical types Integer, Float, and Complex.
        <br/><br/>
    - **3. Boolean literals:** A Boolean literal can have any of the two values: True or False.
        <br/><br/>
    - **4. Special literals:** Python contains one special literal i.e., None. 
        - None is used to specify to that field that is not created. 
        - It is also used for end of lists in Python.
        <br/><br/>
    - **5. Literal Collections:** There are four different literal collections List literals, Tuple literals, Dict literals, and Set literals.
        - Eg. (2, 32, 'xyz', 2.5),[2, 32, 'xyz', 2.5],{'Roll no':2, 'Age':32, 'name':'xyz', 2.5:'avg'}, {2, 32, 'xyz', 2.5}

## Variable or Reference Variable
- A variable is used to store value or data which can later be used in the program.
- A Python variable is an entity which stores the memory address of referenced value(object) stored inside memory.
    - Variable is created inside in stack memory
    - Referenced value is stored inside heap memory.
    - Similar to pointers in C, variables in Python refer to values (or objects) stored somewhere in memory.
- In python, we don't need to declare a variable before using it. Simply assign a value to the variable and it will exist.
- Python is a dynamically typed language hence we need not define the type of the variable while declaring it. The interpreter implicitly binds the value with its type.
<br/><br/>
- In fact, all variable names in Python are said to be references to the values, some of which are front loaded by Python and therefore exist before the name references occur. 
- Python keeps an internal counter on how many references an object has. Once the counter goes to zero - meaning that no reference is made to the object - the garbage collector in Python removes the object , thus freeing up the memory.

## Data Types and Data Structures in python
- Every object in python has a datatype.
- Since everything is an object in Python programming, data types are actually classes and variables are instances (object) of these classes.
- Most common and important types (classes):
    - **Numeric Datatype:** They are defined as int, float and complex class in Python.
    - **String Datatype:** They are defined as str class in Python.
    - **Tuple Datatype:** They are defined as tuple class in Python.
    - **List Datatype:** They are defined as list class in Python.
    - **Dictionary Datatype:** They are defined as dict class in Python.
    - **Sets Datatype:** They are defined as sets class in Python.
    - **Boolean Datatype:** They are defined as bool class in Python.
- We can use the type() function to know which class a variable or a value belongs to and the isinstance() function to check if an object belongs to a particular class.

**NOTE: Data types are those which specify the type of data represented by the variable and literal is the that is stored in to the variable. A literal is a source code representation of a fixed value. They are represented directly in the code without any computation. Literals can be assigned to any primitive type variable.**

- **Variable assignment:**

        - 1. **Single variable assignment**    
            - Ex. a= 100

## Python statements
- Statements are the instructions that a python interpreter can execute.

````python
a = 'Hello World' # single-line statement
print(a) # single-line statement
 
b = 1+2+3\
+4+5+6   # Multi-line statement
print(b)

c = (1+2+3
+4+5+6)
print(c)
````

````python
var1= 10 # var1 variable will refer to a memory having int type object
print(var1)
````

- **Check datatype of object:** type()

````python
print(type(var1))
````

- **Address of object in heap:**

````python
print(id(var1))
````

````python
var2 = var1 # will create another reference variable var2 containing reference to same object as var1
print(var2)
print(id(var2))
````

Hence, python allocate same memory location to variable with same value(Automatic memory allocation)

````python
var1 = var1+1 # new int object 11 will be created and var1 will now refer to that new object
print(var1)
print(id(var1))
````

````python
var3= 10
print(var3)
print(id(var3))
````

- Both var2 and var3 refer to same memory location(i.e, same object) because python optimises the memory utilization by allocating the same object reference to a new variable, if object already exist with the same value.

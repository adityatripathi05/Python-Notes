- **Source file** contains high level instructions weitten by programmer. Extension of this file is ``.py```.


- **Python Compiler** is the integral part of python compiler and checks the syntax by compiling the program. 
    - If there is no syntax error it will generate a intermediate code i.e, ``byte code``` from source file. Extension of this file is ```.pyc```.   
    - If there is syntax error, byte code for the program will not form and the output of the program will be "SyntaxError".
    - Compiled code is run by python interpreter.
    
    
- **Byte code** is the intermediate code which is neither readable by the user nor machine. Byte code converted to machine code everytime when the program is run.


- **PVM (Python Virtual Machine)** is the integral part of python compiler and it generate low level code form python byte code in "on the fly mode" i.e, saved no where. Machine code is directly handover to CPU.
    - Steps in PVM:
        - **Step 1) Laxing:** It is responsible for dividing program into tokens. (Tokens are smallest unit of a program)
        - **Step 2) Parsing:** Tree(Bunch of nodes) formation on which token is placed at each node.
        Parent node and child node
        - **Step 3) Compiler:** Convert high level language to low level language. In python convert unicode code to binary code.
        - **Step 4) Execution:** Operation is performed and output is generated in binary form which is again converted to unicode code and final output on output screen.

**NOTE: After completing the execution of program interpreter delete the .pyc file. Machine code is not saved in interpreter but we can save it by using it as a module**

### Python 2.x Vs Python 3.x:

**1) Division Operator:**
- In python 2.x, the return type of a division (/) operation depends on its operands. We need to use the floating value (i.e, 7.0/5 and -7/5.0) to get desired output (i.e, 1.4 and -1.4 otherwise result will be 1 and -2 i.e, round the calculation to nearest whole numberThe return type of a division (/) operation depends on its operands)
- In python 3.x, division (/) operation always returns a float.

**2) print function:**
- In python 2.x, with print pair of paranthesis is not mendatory(i.e, can be used a property also)
- In python 3.x, it is mendatory to use paranthesis with print i.e, print()

**3) Encoding format:**
- In python 2.x, has implicit ASCII string type, a separate unicode type, but there is no byte type.
- In python 3.x, implicit str type is Unicode and and byte type.

**4) xrange():**
- In python 2.x, has both range and xrange function. When you need to iterate one object at a time, use xrange and when you need an actual list, use range function. xrange is generally faster & saves memory. range() returns a list i.e, range(3) returns [0, 1, 2] while xrange() returns a xrange object i.e, xrange(3) returns iterator object.
- In python 3.x, the range() now does what xrange() does in Python 2.x, so to keep our code portable.

**5) Error Handling:**
- In python 3.x, ```as``` keyword is required.

**6) Input Function:**
- In python 2.x, when using input(), python automatically converts the datatype based on the input. And raw_input() gets the input as text.
- In Python 3.x, the input function acts like raw_input() from Python 2.x and it always returns string type.

**7) Round Function:**
- In python 2.x, round() always results in a floating point number.
- In python 3.x, round() return results in n digit precision.

### Python Identifiers
Identifiers in python are those names that identify a variable, function, class, module, package, or object. It helps in differentiating one entity from another.

- **Python Identifier Writing Rules:**
  - Only lowercase or uppercase letters (a -z, A-Z) or digits (0-9) or an underscore can be combined to form an identifier. For example, myvar, student_Class, employee0, _var1, var_100 are valid identifiers.
  - An identifier cannot be started with a digit or number. For example, myvar9 is valid identifier while 9myvar is not.
  - Python keywords or reserved words are not allowed to be used as identifiers.
  - All the punctuation characters like !, %, &, $, ., cannot be used in identifiers.

### Python Arithmetic Operator:
Arithmetic operators in Python are used to perform arithmetic operations such as addition, subtraction, multiplication, division etc.
1. **+ (Addition):** Adds the operands. **Eg:** x + y


2. **– (Subtraction):** Performs subtraction of right operand from the left operand. **Eg:** x – y


3. ***(Multiplication):** Multiplies two operands. **Eg:** x * y


4. **/ (Division):** Divides the operand at the left side by the operand at the right side. **Eg:** x / y


5. **% (Modulus):** Returns the remainder of a division of the left operand by the right operand. **Eg:** x % y


6. **// (Floor division):** Returns the smallest integer number closest to quotient and also less than the quotient. **Eg:** x // y


7. ****(Exponentiation):** Returns the left-hand operand power right-hand operand. **Eg:** x ** y 

````python
num_1, num_2= 20, 3
````

````python
print(f'Addition of {num_1} and {num_2} is {num_1 + num_2}')

print(f'Subtraction of {num_1} and {num_2} is {num_1 - num_2}')

print(f'Multiplication of {num_1} and {num_2} is {num_1 * num_2}')

print(f'Division of {num_1} and {num_2} is {num_1 / num_2}')

print(format((num_1 / num_2), '.3f'))  # corrected upto 3 decimal places

print(f'Floor Division of {num_1} and {num_2} is {num_1 // num_2}')

print(f'Floor Division of {-num_1} and {num_2} is {-num_1 // num_2}')

print(f'Modulus of {num_1} and {num_2} is {num_1 % num_2}')

print(f'Exponent of {num_1} and {num_2} is {num_1 ** num_2}')
````

**NOTE: Output of division is always in float.**

### Python Assignment Operators:
In python, the operators that are used to assign values to variables are called assignment operators. The major assignment operator is an equal sign (=). 

Assignment operators in python are:
1. **=:** Assigns value at the right side to the variable at left. **Eg:** a = 100, assigns 100 to variable a.


2. **+=:** Adds the operand at the right side to the variable at left and assigns the result to the variable at left. **Eg:** a += 1, equivalent to a = a + 1.


3. **-=:** Subtracts the operand at the right side from the variable at left and assigns the result to the variable at left. **Eg:** a -= 1, equivalent to a = a – 1.


4. ***=:** Multiplies the operand at the right side to the variable at left and assigns the result to the variable at left. **Eg:** a *= 2, equivalent to a = a * 2.


5. **/=:** Divides the variable at the left side by the operand at right and assigns the result to the variable at left. **Eg:** a /= 2, equivalent to a = a / 2.


6. **%=:** Performs modulo operation on two operands and assigns the result to the variable at left. **Eg:** a %= 2, equivalent to a = a % 2.


7. ****=:** Performs an exponential operation on the operands and assigns the result to the variable at left. **Eg:** a ***= 2, equivalent to a = a ** 2.


8. **//=:** Performs floor division on the operands and assigns the resulting value to the left variable. **Eg:** a //= 2, equivalent to a = a // 2.

````python
a = int(input("Enter no. "))
b = int(input("Enter no. "))
````

````python
a += b
print(f"Addition is {a}")
````

````python
a -= b
print(f"Subtraction is {a}")
````

````python
a *= b
print(f"Multiplication is {a}")
````

````python
a /= b
print(f"Division is {a}")
````

````python
a //= b
print(f"Floor division is {a}")
````

````python
a %= b
print(f"Remainder is {a}")
````

````python
a **= b
print(f"Power is {a}")
````

### Python Comparison (Relational) Operators:
The comparison or relational operators in python compare the values or variables on either side of them and return True or False based on the relation.

The comparison operators allowed in python are:
1. **== (Equal to):** Returns True if the operands at both sides are equal, returns False otherwise. **Eg:** a == b.


2. **!= (Not equal to):** Returns True if the operands are not equal, returns False otherwise. **Eg:** a != b.


3. **> (Greater than):** Returns True if the left operand is greater than the operand at right, returns False otherwise. **Eg:** a > b.


4. **>= (Greater than or equal to):** Returns True if the left operand is either greater than or equal to the operand at right, returns False otherwise. **Eg:** a >= b.


5. **< (Less than):** Returns True if the left operand is less than the operand at right, returns False otherwise. **Eg:** a < b.


6. **<= (Less than or equal to):** Returns True if the left operand less than or equal to the operand at right, returns False otherwise. **Eg:** a <= b.



````python
num_1, num_2= 20, 3

print(f'Is {num_1} equal to {num_2}: {num_1 == num_2}')

print(f'Is {num_1} not equal to {num_2}: {num_1 != num_2}')

print(f'Is {num_1} greater than {num_2}: {num_1 > num_2}')

print(f'Is {num_1} greater than or equal to {num_2}: {num_1 >= num_2}')

print(f'Is {num_1} less than {num_2}: {num_1 < num_2}')

print(f'Is {-num_1} less than or equal to {num_2}: {-num_1 <= num_2}')
````

### Python Logical Operators:
The logical operators in python allow us to combine two or more conditional operators and return True or False based on the results of those conditional operators combined with logical operators. 

The logical operators in python are:
1. **and:** Returns True if both the conditions are true otherwise returns False. **Eg:** (a > 10) and (a < 20).


2. **or:** Returns True if one of the conditions is true, otherwise returns False. **Eg:** (a > 10) or (a < 20).


3. **not:** Also called negation, this operator is used to reverse the logical state of its operand. **Eg:** not (a > 10).

````python
a = 11
print((a > 10) and (a < 20))

b = 5
print((b > 10) or (b < 20))

print(not (a > 10))
````

### Python Bitwise Operators:
Bitwise operators perform bit operations bit by bit on the operands. They support one or more operands by converting the operands into binary bits or bit patterns and applying bit operations on individual bits instead of as a whole. 

The bitwise operators in python are:
1. **&:** Performs bitwise AND operation. **Eg:** a & b.
2. **|:** Performs bitwise OR(Inclusive OR) operation. **Eg:** a | b.
3. **~:** Performs bitwise NOT operation(Complement operator). **Eg:** ~ a where ~ is called Tilde symbol.
4. **^:** Performs bitwise XOR(Exclusive OR) operation. **Eg:** a ^ b.
<img src= './Images/1.1 Image b.jpg' width= 700>
5. **>>:** Performs bitwise right shift operation. **Eg:** a >> 1.
6. **<<:** Performs bitwise left shift operation. **Eg:** a << 1.


````python
num_1 = 60            # 60 binary is 0011 1100 
num_2 = 13            # 13 binary is 0000 1101 
print(f"{bin(num_1)} is binary of {num_1}")
print(f"{bin(num_2)} is binary of {num_2}")
````

````python
print(f"Bitwise-AND of {num_1} & {num_2} is {num_1 & num_2}")
print(f"In binary format {bin(num_1)} & {bin(num_2)} is {bin(num_1 & num_2)}")
````

````python
print(f"Bitwise-OR of {num_1} | {num_2} is {num_1 | num_2}")
print(f"In binary format {bin(num_1)} | {bin(num_2)} is {bin(num_1 | num_2)}")
````

````python
print(f"Bitwis-XOR of {num_1} ^ {num_2} is {num_1 ^ num_2}")
print(f"In binary format {bin(num_1)} ^ {bin(num_2)} is {bin(num_1 ^ num_2)}")
````

````python
print(f'Bitwise-NOT of {num_1} is {~ num_1}')
print(f'In binary format, {bin(num_1)} is {bin(~ num_1)}')
````

````python
print(f'Bitwise rightshift of {num_1} is {num_1 >> 2}')
print(f"In binary format {bin(num_1)} rightshifted to {bin(num_1 >> 2)}")
````

````python
print(f'Bitwise leftshift of {num_1} is {num_1 << 2}')
print(f"In binary format {bin(num_1)} leftshifted to {bin(num_1 << 2)}")
````

**NOTE:** Binary of 60 is 0011 1100 then it's complement will be 1100 0011 (just replace 0 with 1 and vice-versa). 

Now 1100 0011 is actually binary of -61. **Let's find out how?**

**We always store positive numbers in system hence to store a negative number we have something called 2's complement (which is nothing but 1's complement+1 ).**

Binary of 61 is 0011 1101 whose 1's complement will be 1100 0010 and 2's complement will be 1100 0011.

### Python Membership Operators (in / not in):
The membership operators in python are used to test for membership of values or variables in sequences like strings, lists, tuples, sets, and dictionaries. 

There are two membership operators available in python as follows:
1. **in:** Returns True if the value or variable is found in the sequence. **Eg:** 10 in a.
2. **not in:** Returns True if the value or variable is not found in the sequence. **Eg:** 10 not in a.

````python
var = 'United Group of Institutions'
print('U' in var)
print('Z' in var)
print('z' not in var)
````

### Python Identity Operators (is / is not):
Python identity operators are used to compare the memory locations of two objects. 

The two identity operators supported by python are as follows:
1. **is:** Returns True if the two objects have the same memory locations. **Eg:** a is b.
2. **is not:** Returns True if the two objects do not have the same memory locations. **Eg:** a is not b.

````python
num1 = 10
num2 =20
print(num1 is num2)
print(num1 is 10)
````

````python
list1 = [1,2,3]
list2 = [1,2,3]
print(list1 is list2)
````

since list, tuple, dictionary are complex order datatype. Hence list1 and list2 are same but not equal hence having different pointer reference.

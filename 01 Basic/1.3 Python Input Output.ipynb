## Method to get output
- **syntax:** ```print(value1,value2, ..., sep=' ', end='\n')```
    - Parameter **'sep'** for seperating values. By default the value of this parameter is ```' '``` i.e, space
    - Parameter **'end'**. By default the value of this parameter is '\n' i.e, the new line character.

````python
print('Ducat', "India")
````

````python
print("India", 'Ducat', sep='\n')
````

````python
print("India", 'Ducat', sep='\n')
print("Python")
print("India", 'Ducat', end='_')
print('Python')
````

## Methods to take input

- **Static input:** 
    - Using assignment operator

````python
name, age= 'Shiva', 23
````

- **Dynamic input of string:**
    - Using **input()**
        - **Syntax:** `input([prompt])` 
            - where prompt/message is optional.
        - **NOTE:** Return type of input() is string type object.

````python
name = input("Enter your name: ")
print(name,'belongs to', type(name))
````

````python
age = input("Enter your age: ")
print(age,'belongs to', type(age))
````

- **For numeric:** Use typecasting

````python
age = int(input('Enter your age: '))
print(age,'belongs to',type(age))
````

````python
perc = float(input('Enter your percentage: '))
print(perc,'belongs to',type(perc))
````

**NOTE: In float typecast we can input integer as conversion from broader datatype to a narrower datatype is possible in python. But conversion from int to string is not possible.**

- **eval():** It automatically evaluate the datatype of value.
    - **NOTE: Using eval(), string must be input inside quotes.**

````python
data = eval(input("Enter anything: "))
print(data,'belongs to',type(data))
````

- **Multiple runtime input:**
    - **split():** This method returns a list of strings after breaking the given string by the specified separator. 
        - **Syntax:** ```str.split(separator, maxsplit)```

````python
name, branch, year = input('Enter your name, branch, year seperated ny comma: ').split(',')
print(f'My name is {name} and my branch is {branch} and presently in {year} year.')
````

## Module in Python:
- All Python programs can call a basic set of functions called built-in functions, including the print(), input(), and len() functions you’ve seen before. These functions are defined inside a **builtins module**. 
- Python also comes with a set of modules called the **standard library**.
- Each module is a Python program that contains a related group of functions that can be embedded in our programs.
- A module is a python file that has collection of identifiers that have been bundled together in a single python file. 
    - i.e, grouping of similar code for easy access.

***
### Advantages of using Module:
- Used for code organization, packaging and reusability
    - **Reusability:** Module can be used in some other python code. Hence it provides the facility of code reusability.
- **Categorization:** Similar type of attributes can be placed in one module.
- **Security**
- **Execution time:** Is fast

- **NOTE: It saves the .pyc file, so that module doesn't run everytime for compilation. It just run .pyc file.**

***
### Types of Module:
- There are two types of module in Python:
    - 1) **Standard Library Modules:** Some of them are as follows: math, random , threading , collections , os , mailbox , string , time , tkinter etc.
        - Ref. <https://docs.python.org/3/py-modindex.html>
    - 2) **User Defined Modules**

### Create a module:
- To create a module just save the code you want in a file with the file extension `.py`.


### Use a module:
- Before we can use the functions in a module, we must load that module in our python code.
- Python provides two types of statements to load a module as defined below:
    - 1. **import statement:** The import statement is used to import all the functionality of one module.
        - **Syntax:** `import module1,module2,........ module n`
    <br/><br/>
    - **NOTE: We can import multiple modules with a single import statement, but a module is loaded once regardless of the number of times, it has been imported into our file.**
    <br/><br/>
    - 2) **from-import statement:** Instead of importing the whole module into the namespace, python provides the flexibility to import only the specific attributes of a module.
        - **Syntax:** `from <module-name> import <name 1>, <name 2>..,<name n>`
<br/><br/>
- The from...import statement is always better to use if we know the attributes to be imported from the module in advance. It doesn't let our code to be heavier.
- **NOTE:** Using from-import statement, we can also import all the attributes from a module by using *
    - `from module_name import *`
- **NOTE: When importing using the from keyword, do not use the module name when referring to elements in the module.**


### Renaming a module:
- We can create an alias when we import a module, by using the `as` keyword.
- **Syntax:** `import <module-name> as <specific-name>` 

### Python Module Attributes:
- Ref. <https://docs.python.org/3/reference/import.html#import-related-module-attributes>

#### `__name__` Attribute: 
- This attribute returns the name of the module. This name is used to uniquely identify the module in the import system.
- By default, the name of the file (excluding the extension .py) is the value of `__name__` attribute.
    - However, value of the `__name__` attribute is set to `__main__` when checked on the interpreter.
    - when the module is imported, its `__name__` is set to its filename.

#### `__doc__` Attribute:
- The `__doc__` attribute denotes the documentation string (docstring) line written in a module code.
     - i.e, `__doc__` attribute will return a string defined at the beginning of the module code.
     
     
#### `__file__` Attribute:
- `__file__` is an optional attribute which holds the name and path of the module file from which it is loaded.


#### `__dict__` Attribute:
- The `__dict__` attribute will return a dictionary object of module attributes, functions and other definitions and their respective values.
- dir() is a built-in function that also returns the list of all attributes and functions in a module.

### Locating Modules:
- When you import a module, the Python interpreter searches for the module in a given list of directories in following sequences−
    - 1) The current directory.
    - 2) If the module isn't found, Python then searches each directory in the shell variable PYTHONPATH.
    - 3) If all else fails, Python checks in the list of installation-dependent default path (e.g., /usr/lib/python).
<br/><br/>    
- The list of directories searched by Python is given by the ```sys.path``` variable.
- Modules must be located in the search path, therefore we can:
    - write our own modules within directories already defined in the search path (e.g. ...local/lib/python3.7/dist-packages).
    - modify the environment variable PYTHONPATH to include the directories containing the user-defined modules.
<br/><br/> 
- **To modify the module path:**
```python
import sys
new_path = <module path>
if new_path not in sys.path:
    sys.path.append(new_path)
```

## random Module:
- This module is used to generate random numbers.
- <https://docs.python.org/3/library/random.html>


- **random()** : Returns random floating point number in the range \[0.0, 1.0).


- **randint(a,b)** : Returns a random integer N in the range \[a,b\].


- **randrange(start,stop,step)** : Returns a randomly selected element from range(start, stop, step). 


- **choice(seq)** : Returns a random element from the non-empty sequence seq.


- **choices(population, weights=None, cum_weights=None, k=1)** : Returns a k sized list of elements chosen from the population __with replacement__. If a weights sequence is specified, selections are made according to the relative weights.


- **shuffle(seq)** : Shuffle the sequence seq in place.


- **sample(population, k)** : Return a k length list of unique elements chosen from the population sequence __without replacement__.

````python
from random import random,randint,randrange,choice,shuffle,seed
````

````python
#random.random(): Generates a random float number between 0.0 to 1.0(where 1.0 is exclusive). The function doesn't need any arguments. 
print(random())
````

````python
#random.randint(x,y): Returns a random integer between the specified integers, where both the numbers are inclusive.
print(randint(1,10))
````

````python
# random.randrange(): Returns a randomly selected element from the range created by the start, stop and step arguments. 
# The value of start is 0 by default. Similarly, the value of step is 1 by default.
print(randrange(1,10))
print(randrange(1,10,2))
print(randrange(0,101,10))
````

````python
#random.choice(): Returns a randomly selected element from a non-empty sequence. 
#An empty sequence as argument raises an IndexError.
print(choice('computer'))
print(choice([12,23,45,67,65,43]))
````

````python
#random.shuffle(): functions randomly reorders the elements in a list.
l = [10,20,30,40]
print(l)
shuffle(l)
print(l)
shuffle(l)
print(l)
# random uses system clock to generate or shuffle random numbers.
````

````python
#random.seed(): gernerate a specific pattern of generating random number and is fixed for a particular seed value
seed(1)  
l = [10,20,30,40]
shuffle(l)
print(l)
seed(2)
shuffle(l)
print(l)
````

## collections Module:
- This module implements **specialized container datatypes** providing alternatives to Python’s general purpose built-in containers like dict, list, set, and tuple.
- Ref: <https://docs.python.org/3/library/collections.html>

- Some important container datatypes are:

    - **Counter** : for counting hashable objects. It is an unordered collection where elements are stored as dictionary keys and their counts are stored as dictionary values.

    - **ChainMap** : for quickly linking a number of mappings so they can be treated as a single unit

    - **defaultdict** : calls a factory function to supply values for missing keys in a dictionary.

    - **OrderedDict** : a dict subclass which remembers the order in which entries were added

````python
from collections import *
````

#### namedtuple()
- Returns a tuple-like object with named fields. These field attributes are accessible by lookup as well as by index.
- **Syntax:** `collections.namedtuple(type_name, field-list)`

````python
stu= namedtuple('Student', ['name', 'age', 'marks'])
#statement declares a Student class having name, age and marks as fields.
````

````python
#create a new object of this namedtuple, do the following:
s1=stu("Imran", 21, 98)

# access value of field by attribute lookup
print(s1.name)

# access value of field by index
print(s1[0])
````

#### OrderedDict()
- The OrderedDict() function is similar to a normal dictionary object in Python. 
- However, it remembers the order of the keys in which they were first inserted.

````python
d1= OrderedDict()
d1['A']=65
d1['C']=67
d1['B']=66
d1['D']=68

for k,v in d1.items():
    print (k,v)
````

#### deque()
- A deque object support appends and pops from either ends of a list. 
- It is more memory efficient than a normal list object. 
- In a normal list object, the removal of any item causes all items to the right to be shifted towards left by one index. Hence, it is very slow.

````python
q= deque([10,20,30,40])
q.appendleft(0)
print(q)
````

````python
deque([0, 10, 20, 30, 40])
q.append(50)
print(q)
````

````python
deque([0, 10, 20, 30, 40, 50])
q.pop()
print(q)
````

````python
deque([0, 10, 20, 30, 40])
q.popleft()
print(q)
````

### Time Module:
- Python time module provides many ways of representing time in code, such as objects, numbers, and strings.
- It also provides functionality other than representing time, like waiting during code execution and measuring the efficiency of your code.

````python
import time
print(time.__doc__)
````

````python
print(dir(time))
````

````python
#time.time(): Measures the passage of time(in sec) since epoch time (January 1, 1970 UTC, Coordinated Universal Time)
print(time.time()) # in sec
print(time.time_ns()) # in nano sec
````

````python
# time.asctime(): Convert the Python time, represented as the number of elapsed seconds since the epoch, to a string.
print(time.asctime())
````

- The string representation of time, also known as a **timestamp**, returned by ctime() is formatted with the following structure:
    - Day of the week: Mon (Monday)
    - Month of the year: Feb (February)
    - Day of the month: 25
    - Hours, minutes, and seconds using the 24-hour clock notation: 19:11:59
    - Year: 2019

````python
#time.localtime(): Providelocal time from the number of seconds elapsed since the epoch
t= time.localtime()
print(t)
````

````python
print(f'{t[2]}/{t[1]}/{t[0]}')
````

````python
#struct_time: It utilizes NamedTuple, from Python’s collections module, to associate the tuple’s sequence of numbers with useful identifiers
time_tup=(2020,3,26,16,17,55,3,86,0)
print(time.struct_time(time_tup))
````

````python
#time.sleep(n): Pauses the process thread for n seconds
print('Welcome')
time.sleep(3)
print('Welcome again after 3 seconds')
````

## Pickle Module:
- It is a way to save python data objects & next time when we will need it ,we can simply load it up & get back data objects. This process known as object serialization.
- **pickle.dump()** is the method for saving the data to the pickle file with .p & .pkl extension.
- **pickle.load()** is the method for loading pickle file.

````python
import pickle
print(pickle.__doc__)
````

````python
# Save a pickle file
my_dict = {'name':'ravi', 'age':24, 1:'app'}
with open('test.pkl', 'wb') as f:
    pickle.dump(my_dict, f)
````

````python
# Load a pickle file
with open('test.pkl' , 'rb') as f:
    ndict = pickle.load(f)

print(ndict)
````
